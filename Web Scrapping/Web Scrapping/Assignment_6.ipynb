Use Selenium to navigate to the Zomato website, wait for the search bar to load, and extract the placeholder text from the search input field.<br><br><em><strong>Hint:</strong> Use WebDriverWait with expected_conditions to wait for the element to be present before extracting its attribute.</em>

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

# Initialize the Chrome browser instance
driver = webdriver.Chrome()

try:
    # Navigate to the Zomato homepage
    url = "https://www.zomato.com"
    print(f"Opening {url}...")
    driver.get(url)

    # Initialize explicit wait (wait up to 10 seconds)
    wait = WebDriverWait(driver, 10)

    # Wait until the search input element is present in the DOM and visible
    # Zomato's search bars typically contain 'Search' in their placeholder text
    search_input = wait.until(
        EC.presence_of_element_located((By.CSS_SELECTOR, "input[placeholder*='Search']"))
    )

    # Extract the value of the 'placeholder' attribute
    placeholder_text = search_input.get_attribute("placeholder")

    print("\nSuccessfully located the search bar!")
    print(f"Placeholder Text: '{placeholder_text}'")

except Exception as e:
    print(f"An error occurred while interacting with the page: {e}")

finally:
    # Ensure the browser window closes to free up system resources
    driver.quit()
```

Write a Selenium script to open the BookMyShow homepage, scroll down to load more content dynamically, and print the names of the first 5 movies displayed.<br><br><em><strong>Hint:</strong> Use driver.execute_script('window.scrollTo(...)') to scroll, and find_elements to get movie names.</em>

```python
import time
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

# Initialize the Chrome browser instance
driver = webdriver.Chrome()

try:
    # Navigate directly to BookMyShow's movies exploration page
    url = "https://in.bookmyshow.com/explore/movies"
    print(f"Opening {url}...")
    driver.get(url)

    # Initialize explicit wait
    wait = WebDriverWait(driver, 10)

    # Wait until headings or movie cards are present in the DOM
    wait.until(EC.presence_of_element_located((By.TAG_NAME, "h3")))

    # Scroll down the page using JavaScript to trigger dynamic content loading
    print("Scrolling down to load dynamic content...")
    driver.execute_script("window.scrollTo(0, 1000);")
    
    # Pause briefly to allow lazy-loaded images and text elements to render
    time.sleep(3)

    # Use find_elements to locate movie title heading elements (typically stored in <h3> tags)
    movie_elements = driver.find_elements(By.TAG_NAME, "h3")

    print("\n--- First 5 Movies Discovered ---")
    count = 0
    for elem in movie_elements:
        title = elem.text.strip()
        # Filter out empty strings or non-movie headings if any exist
        if title and count < 5:
            count += 1
            print(f"{count}. {title}")
            
    if count == 0:
        print("No movie titles found. (BookMyShow's layout or class structure may have changed).")

except Exception as e:
    print(f"An error occurred during automation: {e}")

finally:
    # Always close the browser session to release system resources
    driver.quit()
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

# Initialize the Chrome browser instance
driver = webdriver.Chrome()

try:
    # 1. Navigate to the demo login page
    url = "https://www.saucedemo.com/"
    print(f"Opening {url}...")
    driver.get(url)

    # Initialize explicit wait
    wait = WebDriverWait(driver, 10)

    # 2. Locate input fields and login button using stable IDs
    username_field = wait.until(EC.presence_of_element_located((By.ID, "user-name")))
    password_field = driver.find_element(By.ID, "password")
    login_button = driver.find_element(By.ID, "login-button")

    # 3. Enter credentials and submit
    print("Entering credentials...")
    username_field.send_keys("standard_user")
    password_field.send_keys("secret_sauce")
    login_button.click()

    # 4. Wait for the post-login welcome header/element to load
    print("Waiting for dashboard to load...")
    welcome_header = wait.until(
        EC.presence_of_element_located((By.CSS_SELECTOR, "span.title"))
    )

    # 5. Extract and print the message
    print("\n--- Automation Output ---")
    print("Login successful!")
    print(f"Welcome / Page Header: '{welcome_header.text}'")

except Exception as e:
    print(f"An error occurred during automation: {e}")

finally:
    # Always close the browser session to free up system resources
    driver.quit()
    print("\nBrowser session closed successfully.")
```
