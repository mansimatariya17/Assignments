Perform univariate analysis on the 'runs' column: calculate and print the mean, median, mode, minimum, and maximum runs scored by batsmen in the IPL_2023_Batsmen dataset.

```python
print('Mean Runs:', round(df['Runs'].mean(), 2))
print('Median Runs:', df['Runs'].median())
print('Mode Runs:', df['Runs'].mode().tolist())
print('Minimum Runs:', df['Runs'].min())
print('Maximum Runs:', df['Runs'].max())
```

Create a Pandas DataFrame for your top 5 favorite Zomato restaurants with columns: 'name', 'rating', 'votes', and 'avg_cost'. Perform bivariate analysis between 'rating' and 'votes' by calculating their correlation coefficient.

```python
data = {
    'name': ["La Pino's Pizza", 'Barbeque Nation', 'The Belgian Waffle Co.', 'Natural Ice Cream', "McDonald's"],
    'rating': [4.3, 4.6, 4.5, 4.7, 4.2],
    'votes': [2850, 4200, 1950, 3100, 5400],
    'avg_cost': [400, 1600, 300, 250, 350]
}

df1 = pd.DataFrame(data)

corr_value = df1['rating'].corr(df1['votes'])

print(f"Correlation between Rating and Votes: {corr_value:.4f}")
```
