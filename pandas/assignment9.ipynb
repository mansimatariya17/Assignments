Create a Pandas Series of 5 random date strings in the format 'DD-MM-YYYY', then convert this Series to datetime objects using pd.to_datetime and print the result.

```python
date_strings = pd.Series(['15-08-2026', '26-01-2026', '02-10-2026', '12-05-2026', '31-12-2026'])

date_objects = pd.to_datetime(date_strings, format='%d-%m-%Y')

date_objects
```

Given a DataFrame with a 'release_date' column containing movie release dates (as strings), add three new columns: 'year', 'month', and 'day' by extracting these values from the 'release_date' column using Pandas.<br><br><em><strong>Hint:</strong> Use the .dt accessor after converting to datetime.</em>

```python
data = {
    'Movie': ['Pathaan', 'Jawan', 'Animal', 'Kantara 2', 'Stree 2'],
    'release_date': ['2023-01-25', '2023-09-07', '2023-12-01', '2024-10-02', '2024-08-15']
}

df = pd.DataFrame(data)

df['release_date'] = pd.to_datetime(df['release_date'])

df['year'] = df['release_date'].dt.year
df['month'] = df['release_date'].dt.month
df['day'] = df['release_date'].dt.day

df
```

Suppose you have a DataFrame of Zomato Gold subscriptions with columns 'user_id' and 'start_date' (as strings). Calculate the tenure (in days) for each user by subtracting the 'start_date' from today's date, and add it as a new column 'tenure_days'.<br><br><em><strong>Hint:</strong> Use pd.to_datetime for conversion and (today - start_date).dt.days for calculation.</em>

```python
data = {
    'user_id': ['ZOM101', 'ZOM102', 'ZOM103', 'ZOM104', 'ZOM105'],
    'start_date': ['2026-01-15', '2025-06-20', '2026-08-01', '2024-12-10', '2026-09-15']
}

df = pd.DataFrame(data)
df['start_date'] = pd.to_datetime(df['start_date'])

today = pd.Timestamp.now().normalize()

df['tenure_days'] = (today - df['start_date']).dt.days

df
```
