Create a scatter plot showing 'amount spent on Zomato orders' vs 'number of orders placed' for 12 friends, and label each point with the friend's name.<br><br><em><strong>Hint:</strong> Use the 'annotate' function in matplotlib to add labels.</em>

```python
names = ["Aarav", "Priya", "Rahul", "Neha", "Riya", "Karan","Ananya", "Vivek", "Sneha", "Arjun", "Meera", "Rohan"]
amount_spent = [500, 800, 300, 1200, 650, 1500,400, 1000, 750, 1800, 550, 1300]
number_of_orders = [5, 8, 3, 12, 6, 15,4, 10, 7, 18, 5, 13]

plt.figure(figsize=(10,6))
plt.scatter(number_of_orders, amount_spent, color="#4944CE")

for i in range(len(names)):
    plt.annotate(
        names[i],
        (number_of_orders[i], amount_spent[i]),
        xytext=(5, 5),
        textcoords="offset points"
    )
plt.tight_layout()   
plt.show()
```

Add a trend line to your Zomato scatter plot by calculating the best-fit line using numpy's polyfit(), then plot it on top of the scatter plot.<br><br><em><strong>Hint:</strong> Use np.polyfit() to get the slope and intercept, then plot the line with plt.plot().</em>

```python
names = ["Aarav", "Priya", "Rahul", "Neha", "Riya", "Karan","Ananya", "Vivek", "Sneha", "Arjun", "Meera", "Rohan"]
amount_spent = [500, 800, 300, 1200, 650, 1500,400, 1000, 750, 1800, 550, 1300]
number_of_orders = [5, 8, 3, 12, 6, 15,4, 10, 7, 18, 5, 13]

plt.figure(figsize=(10, 6))
plt.scatter(number_of_orders, amount_spent,
            color="purple", s=80, label="Friends")

for i in range(len(names)):
    plt.annotate(
        names[i],
        (number_of_orders[i], amount_spent[i]),
        xytext=(5, 5),
        textcoords="offset points"
    )

slope, intercept = np.polyfit(number_of_orders, amount_spent, 1)

x_line = np.linspace(min(number_of_orders), max(number_of_orders), 100)
y_line = slope * x_line + intercept

plt.plot(x_line, y_line, color="red", linestyle="--",
         label="Best-fit trend line")

plt.title("Zomato Spending vs Number of Orders")
plt.xlabel("Number of Orders Placed")
plt.ylabel("Amount Spent on Zomato (₹)")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()

print("Slope:", slope)
print("Intercept:", intercept)
```

Given a dataset of 'YouTube watch hours' and 'Spotify listening hours' for 15 users, use numpy polyfit to fit a regression line, plot both the scatter and the trend line, and print the equation of the line on the plot.

```python
youtube_hours = [2, 5, 3, 8, 6, 10, 4, 12, 7, 9, 11, 3, 6, 14, 8]
spotify_hours = [4, 6, 5, 9, 7, 12, 5, 14, 8, 10, 13, 4, 8, 16, 9]

# Fit regression line
slope, intercept = np.polyfit(youtube_hours, spotify_hours, 1)

# Calculate trend line values
x_line = np.linspace(min(youtube_hours), max(youtube_hours), 100)
y_line = slope * x_line + intercept

# Plot scatter points
plt.scatter(youtube_hours, spotify_hours,
            color="purple", label="Users")

# Plot regression line
plt.plot(x_line, y_line, color="red", linestyle="--",
         label="Regression Line")

# Display equation on the plot
equation = f"y = {slope:.2f}x + {intercept:.2f}"
plt.text(0.05, 0.95, equation,
         transform=plt.gca().transAxes,
         fontsize=12, color="red", verticalalignment="top")

# Add title and labels
plt.title("YouTube Watch Hours vs Spotify Listening Hours")
plt.xlabel("YouTube Watch Hours per Week")
plt.ylabel("Spotify Listening Hours per Week")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()

# Print regression equation
print("Regression equation:", equation)
```
