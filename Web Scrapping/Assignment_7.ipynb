Use Selenium to open https://www.myntra.com/ in a Chrome browser, locate the search box, and enter the text 'shoes' using send_keys(), then submit the search.

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

# Initialize the Chrome browser instance
driver = webdriver.Chrome()

try:
    # 1. Open the Myntra homepage
    url = "https://www.myntra.com/"
    print(f"Opening {url}...")
    driver.get(url)

    # Initialize explicit wait
    wait = WebDriverWait(driver, 15)

    # 2. Locate the search box using its standard class name on Myntra
    print("Locating search input field...")
    search_box = wait.until(
        EC.presence_of_element_located((By.CSS_SELECTOR, "input.desktop-searchBar"))
    )

    # 3. Enter 'shoes' using send_keys()
    print("Entering 'shoes' into the search bar...")
    search_box.send_keys("shoes")

    # 4. Submit the search by simulating the 'Enter' key press
    print("Submitting search...")
    search_box.send_keys(Keys.RETURN)

    # Wait for the search results page to load and verify the URL changed
    wait.until(EC.url_contains("q=shoes"))
    print("\nSearch submitted successfully!")
    print(f"Current Results URL: {driver.current_url}")

except Exception as e:
    print(f"An error occurred during automation: {e}")

finally:
    # Always close the browser session to release system resources
    driver.quit()
    print("Browser session closed.")
```

Write a Selenium script that clicks the 'Login' button on https://www.zomato.com/ and handles the popup by closing it after it appears.<br><br><em><strong>Hint:</strong> Wait for the popup using WebDriverWait and locate the close button before clicking.</em>

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

# Initialize the Chrome browser instance
driver = webdriver.Chrome()

try:
    # 1. Open the Zomato homepage
    url = "https://www.zomato.com/"
    print(f"Opening {url}...")
    driver.get(url)
    
    # Maximize window to ensure header elements are fully visible
    driver.maximize_window()

    # Initialize explicit wait
    wait = WebDriverWait(driver, 15)

    # 2. Locate and click the 'Log in' button in the header
    print("Locating the 'Log in' button...")
    login_button = wait.until(
        EC.element_to_be_clickable((By.XPATH, "//a[normalize-space()='Log in']"))
    )
    login_button.click()
    print("'Log in' button clicked successfully.")

    # 3. Wait for the login modal popup to appear in the DOM
    print("Waiting for the login popup modal to render...")
    modal = wait.until(
        EC.presence_of_element_located((By.CSS_SELECTOR, "div[role='dialog']"))
    )

    # 4. Locate and click the close button (typically an SVG or 'X' icon inside the modal header)
    print("Locating and clicking the popup close button...")
    close_button = wait.until(
        EC.element_to_be_clickable((By.CSS_SELECTOR, "div[role='dialog'] svg"))
    )
    close_button.click()
    
    print("\nPopup modal closed successfully!")

except Exception as e:
    print(f"An error occurred during automation: {e}")

finally:
    # Always close the browser session to release system resources
    driver.quit()
    print("Browser session closed.")
```

On https://www.flipkart.com/, use Selenium to close the initial login popup, then click on the 'Mobiles' menu item to navigate to the mobiles section.

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

# Initialize the Chrome browser instance
driver = webdriver.Chrome()

try:
    # Maximize window to ensure the top navigation menu is fully rendered and clickable
    driver.maximize_window()

    # 1. Open the Flipkart homepage
    url = "https://www.flipkart.com/"
    print(f"Opening {url}...")
    driver.get(url)

    # Initialize explicit wait
    wait = WebDriverWait(driver, 10)

    # 2. Close the initial login popup if it appears
    try:
        print("Checking for initial login popup...")
        # Flipkart login modal close button is typically an '✕' button or has specific classes
        close_btn = wait.until(
            EC.element_to_be_clickable((By.XPATH, "//button[contains(@class, '_2KpZ6l') and contains(@class, '_2doB4z')] | //span[text()='✕']/.."))
        )
        close_btn.click()
        print("Login popup closed successfully.")
    except Exception:
        print("No login popup detected (or it was not present). Continuing...")

    # 3. Locate and click the 'Mobiles' menu item in Flipkart's top category bar
    print("Locating the 'Mobiles' menu item...")
    mobiles_menu = wait.until(
        EC.element_to_be_clickable((By.XPATH, "//span[normalize-space()='Mobiles'] | //div[normalize-space()='Mobiles']"))
    )
    mobiles_menu.click()
    print("'Mobiles' menu clicked successfully!")

    # 4. Wait for the URL to update to the mobiles section
    wait.until(EC.url_contains("mobile"))
    print(f"\nSuccessfully navigated to Mobiles section!")
    print(f"Current URL: {driver.current_url}")

except Exception as e:
    print(f"An error occurred during automation: {e}")

finally:
    # Always close the browser session to release system resources
    driver.quit()
    print("Browser session closed.")
```

Build a Selenium script to scroll down the Instagram Explore page (https://www.instagram.com/explore/) multiple times, loading more posts each time, and print the total number of post links collected after each scroll.<br><br><em><strong>Hint:</strong> Use execute_script('window.scrollTo(0, document.body.scrollHeight)') and time.sleep() between scrolls.</em>

```python
import time
from selenium import webdriver
from selenium.webdriver.common.by import By

# Configure Chrome options to reduce basic bot-detection flags
options = webdriver.ChromeOptions()
options.add_argument("--disable-blink-features=AutomationControlled")
options.add_experimental_option("excludeSwitches", ["enable-automation"])
options.add_experimental_option('useAutomationExtension', False)

# Initialize the Chrome browser instance
driver = webdriver.Chrome(options=options)

try:
    # 1. Navigate to Instagram Explore page
    url = "https://www.instagram.com/explore/"
    print(f"Opening {url}...")
    driver.get(url)
    
    # Maximize window to ensure optimal grid rendering
    driver.maximize_window()

    # Pause to allow initial content to load
    time.sleep(4)

    # Use a set to store unique post URLs and prevent duplicates
    collected_links = set()
    total_scrolls = 5  # Adjust this number to scroll more or fewer times

    for scroll_num in range(1, total_scrolls + 1):
        print(f"\n--- Scroll Iteration {scroll_num} of {total_scrolls} ---")

        # 2. Extract all anchor elements currently present in the DOM
        link_elements = driver.find_elements(By.TAG_NAME, "a")

        # Filter and collect links that point to posts (/p/) or reels (/reel/)
        new_links_found = 0
        for elem in link_elements:
            href = elem.get_attribute("href")
            if href and ("/p/" in href or "/reel/" in href):
                if href not in collected_links:
                    collected_links.add(href)
                    new_links_found += 1

        print(f"New unique posts found in this scroll: {new_links_found}")
        print(f"Total unique post links collected so far: {len(collected_links)}")

        # 3. Scroll down to the bottom of the page to trigger infinite loading
        print("Scrolling down to load more posts...")
        driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")

        # 4. Pause to give the dynamic page time to fetch and render new posts
        time.sleep(3)

    print(f"\nScraping complete! Total unique post links gathered: {len(collected_links)}")

except Exception as e:
    print(f"An error occurred during automation: {e}")

finally:
    # Always close the browser session to release system resources
    driver.quit()
    print("Browser session closed.")
```

```python
import time
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

# Configure Chrome options to bypass basic bot-detection flags
options = webdriver.ChromeOptions()
options.add_argument("--disable-blink-features=AutomationControlled")
options.add_experimental_option("excludeSwitches", ["enable-automation"])
options.add_experimental_option('useAutomationExtension', False)
options.add_argument("start-maximized")

# Initialize the Chrome browser instance
driver = webdriver.Chrome(options=options)

try:
    # 1. Navigate to Nykaa lipstick search results page
    url = "https://www.nykaa.com/search/result/?q=lipstick"
    print(f"Opening {url}...")
    driver.get(url)

    # Initialize explicit wait
    wait = WebDriverWait(driver, 15)

    # Use a set to store unique product names and avoid duplicates across scrolls
    product_names = set()
    target_count = 30
    max_scrolls = 10
    scroll_count = 0

    print(f"Scrolling and collecting at least {target_count} product names...\n")

    while len(product_names) < target_count and scroll_count < max_scrolls:
        # Locate product name elements on the current view
        # Nykaa product title classes typically use anchor tags or divs inside product cards
        elements = driver.find_elements(By.CSS_SELECTOR, "div.css-xz75vw, a.css-qluxh5 div, .product-name")
        
        for elem in elements:
            name = elem.text.strip()
            if name and name not in product_names:
                product_names.add(name)

        print(f"Scroll {scroll_count + 1}: Collected {len(product_names)} unique products so far...")

        if len(product_names) >= target_count:
            break

        # Scroll down to the bottom of the page to trigger infinite loading
        driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
        
        # Pause to allow asynchronous AJAX requests to load new product cards
        time.sleep(3)
        scroll_count += 1

    # Convert set to a list and slice to get the first 30
    final_products = list(product_names)[:target_count]

    print(f"\nSuccessfully collected {len(final_products)} product names:\n")
    for index, name in enumerate(final_products, start=1):
        print(f"{index}. {name}")

except Exception as e:
    print(f"An error occurred during automation: {e}")

finally:
    # Always close the browser session to release system resources
    driver.quit()
    print("\nBrowser session closed.")
```
