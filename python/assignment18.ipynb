<p>Use `re.sub()` to mask every phone number in the text as 'XXXXXX' plus its last 4 digits, and save the result as 'fest_messages_masked.txt'. Hint: Capture the last 4 digits in a group and use `r'XXXXXX\1'` in the replacement.</p>

```python
import re

with open("fest_messages.txt", "r", encoding="utf-8") as file:
    text = file.read()

# 1. Extract valid email addresses
email_pattern = r"[\w.+-]+@[\w-]+\.[\w.]+"
emails = re.findall(email_pattern, text)
print(f"Total matched emails: {len(emails)}")

# 2. Extract and count 10-digit mobile numbers (starting with 6-9)
phone_pattern = r"\b[6-9]\d{9}\b"
phones = re.findall(phone_pattern, text)
print(f"Total matched mobile numbers: {len(phones)}")

# 3. Find the first matching date (DD/MM/YYYY) and its line number
date_pattern = r"\d{2}/\d{2}/\d{4}"
for line_num, line in enumerate(text.splitlines(), start=1):
    match = re.search(date_pattern, line)
    if match:
        print(f"First matching date: {match.group()} (found on line {line_num})")
        break

# 4. Mask every phone number as 'XXXXXX' + last 4 digits and save to file
mask_pattern = r"\b[6-9]\d{5}(\d{4})\b"
masked_text, masked_count = re.subn(mask_pattern, r"XXXXXX\1", text)

with open("fest_messages_masked.txt", "w", encoding="utf-8") as out_file:
    out_file.write(masked_text)

print(f"Masked {masked_count} phone numbers and saved to fest_messages_masked.txt")
```

<p>Finish by extracting every amount written like 'Rs 250' with `re.findall(r'Rs\s?(\d+)', text)`, converting to `int`, and printing the total ticket revenue and count. Save the counts of emails, phones and the revenue total in 'fest_summary.json' and run Restart & Run All.</p>

```python
import json
import re

with open("fest_messages.txt", "r", encoding="utf-8") as file:
    text = file.read()

# 1. Extract valid email addresses
email_pattern = r"[\w.+-]+@[\w-]+\.[\w.]+"
emails = re.findall(email_pattern, text)
print(f"Total matched emails: {len(emails)}")

# 2. Extract and count 10-digit mobile numbers (starting with 6-9)
phone_pattern = r"\b[6-9]\d{9}\b"
phones = re.findall(phone_pattern, text)
print(f"Total matched mobile numbers: {len(phones)}")

# 3. Find the first matching date (DD/MM/YYYY) and its line number
date_pattern = r"\d{2}/\d{2}/\d{4}"
for line_num, line in enumerate(text.splitlines(), start=1):
    match = re.search(date_pattern, line)
    if match:
        print(f"First matching date: {match.group()} (found on line {line_num})")
        break

# 4. Mask every phone number as 'XXXXXX' + last 4 digits and save to file
mask_pattern = r"\b[6-9]\d{5}(\d{4})\b"
masked_text, masked_count = re.subn(mask_pattern, r"XXXXXX\1", text)

with open("fest_messages_masked.txt", "w", encoding="utf-8") as out_file:
    out_file.write(masked_text)

print(f"Masked {masked_count} phone numbers and saved to fest_messages_masked.txt")

# 5. Extract 'Rs <amount>' entries, convert to int, and compute ticket revenue
amount_strings = re.findall(r"Rs\s?(\d+)", text)
amounts = [int(val) for val in amount_strings]
total_revenue = sum(amounts)

print(f"Ticket entries found: {len(amounts)}")
print(f"Total ticket revenue: Rs {total_revenue}")

# 6. Save summary counts and revenue total to fest_summary.json
summary = {
    "email_count": len(emails),
    "phone_count": len(phones),
    "total_ticket_revenue": total_revenue,
}

with open("fest_summary.json", "w", encoding="utf-8") as json_file:
    json.dump(summary, json_file, indent=2)

print("Saved summary to fest_summary.json")
```
