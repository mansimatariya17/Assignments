Given a Flipkart-style product reviews dataset with some missing 'rating' values, use pandas to fill all missing ratings with the mean rating of the dataset and print the updated DataFrame.<br><br><em><strong>Hint:</strong> Use the fillna() method with the mean() function.</em>

```python
df = pd.DataFrame({
    'product': ['Phone', 'Laptop', 'Headphones', 'Watch', 'Tablet'],
    'rating': [4.5, 5.0, None, 3.5, None]
})

df['rating'] = df['rating'].fillna(df['rating'].mean())
df
```

You have a Spotify playlist CSV with missing 'duration' values for some songs. Replace all missing durations with the median duration of the available songs using pandas, then save the cleaned DataFrame to a new CSV file.

```python
df = pd.read_csv("C:/Users/Windows/OneDrive/Desktop/Pandas Dataset/spotify_playlist.csv")
df['duration'] = df['duration'].fillna(df['duration'].median())
df
```

Use ChatGPT to generate pandas code that replaces all missing values in a 'followers' column (from a simulated Instagram user data CSV) with 0, then test the code on your own sample data and submit both the code and the result.<br><br><em><strong>Hint:</strong> Ask ChatGPT: 'How do I replace missing values in a pandas DataFrame column with zero?'</em>

```python
data = {
    'username': ['user_01', 'user_02', 'user_03', 'user_04', 'user_05'],
    'followers': [1200, None, 850, None, 2300]
}

df = pd.DataFrame(data)

print("Before replacing missing values:")
print(df)

# Replace missing values in 'followers' with 0
df['followers'] = df['followers'].fillna(0)

print("\nAfter replacing missing values:")
print(df)
```
