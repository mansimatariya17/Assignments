Use the requests library to fetch the HTML content of the Flipkart homepage (https://www.flipkart.com) and print the first 500 characters of the response text.

```python
import requests

# Flipkart URL
url = "https://www.flipkart.com"

# Add headers to mimic a real browser request (prevents 403 Forbidden errors)
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                  "AppleWebKit/537.36 (KHTML, like Gecko) "
                  "Chrome/120.0.0.0 Safari/537.36"
}

try:
    # Send GET request to Flipkart
    response = requests.get(url, headers=headers)
    
    # Raise an exception for HTTP error codes (like 403 or 404)
    response.raise_for_status()
    
    # Print the first 500 characters of the HTML response text
    print("--- First 500 Characters of Flipkart HTML ---")
    print(response.text[:500])

except requests.exceptions.RequestException as e:
    print(f"An error occurred: {e}")
```

Parse the HTML content of https://www.myntra.com using BeautifulSoup and print the title of the page.

```python
import requests
from bs4 import BeautifulSoup

# Target URL
url = "https://www.myntra.com"

# Headers to mimic a real browser request
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                  "AppleWebKit/537.36 (KHTML, like Gecko) "
                  "Chrome/120.0.0.0 Safari/537.36"
}

try:
    # Send GET request to Myntra
    response = requests.get(url, headers=headers)
    response.raise_for_status()
    
    # Parse the HTML content using BeautifulSoup
    soup = BeautifulSoup(response.text, 'html.parser')
    
    # Find and print the title of the page
    if soup.title:
        print("Page Title:", soup.title.string.strip())
    else:
        print("No title tag found on the page.")

except requests.exceptions.RequestException as e:
    print(f"An error occurred: {e}")
```

Extract and display all the unique class names used in <div> tags from the Zomato homepage (https://www.zomato.com) using BeautifulSoup.<br><br><em><strong>Hint:</strong> Use soup.find_all('div') and access the 'class' attribute.</em>

```python
import requests
from bs4 import BeautifulSoup

# Target URL
url = "https://www.zomato.com"

# Headers to mimic a real browser request and prevent blocking
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                  "AppleWebKit/537.36 (KHTML, like Gecko) "
                  "Chrome/120.0.0.0 Safari/537.36"
}

try:
    # Fetch the webpage
    response = requests.get(url, headers=headers)
    response.raise_for_status()
    
    # Parse the HTML content
    soup = BeautifulSoup(response.text, 'html.parser')
    
    # Use a set to store unique class names
    unique_classes = set()
    
    # Find all <div> tags
    div_tags = soup.find_all('div')
    
    for div in div_tags:
        # div.get('class') returns a list of classes if present, or None if no class attribute exists
        class_list = div.get('class')
        
        if class_list:
            # Add each individual class name to the set
            for cls in class_list:
                unique_classes.add(cls)
                
    # Display results
    print(f"Total unique <div> class names found: {len(unique_classes)}\n")
    print("Sample of unique class names:")
    
    # Print sorted sample of up to 25 unique classes for clarity
    for cls in sorted(list(unique_classes))[:25]:
        print(f" - {cls}")

except requests.exceptions.RequestException as e:
    print(f"An error occurred: {e}")
```

Pick any public webpage of an app you use (like BookMyShow, Paytm, or IRCTC), fetch its HTML using requests, and use BeautifulSoup to find and print all elements with an id attribute.

```python
import requests
from bs4 import BeautifulSoup

# Target URL
url = "https://www.zomato.com"

# Headers to mimic a real browser request and prevent blocking
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                  "AppleWebKit/537.36 (KHTML, like Gecko) "
                  "Chrome/120.0.0.0 Safari/537.36"
}

try:
    # Fetch the webpage HTML
    response = requests.get(url, headers=headers)
    response.raise_for_status()
    
    # Parse the HTML using BeautifulSoup
    soup = BeautifulSoup(response.text, 'html.parser')
    
    # Find all elements that possess an 'id' attribute using id=True
    elements_with_id = soup.find_all(id=True)
    
    print(f"Total elements with an 'id' attribute found: {len(elements_with_id)}\n")
    print("Sample of elements with IDs (Tag & ID value):")
    
    # Loop through and display the tag name and its corresponding id value
    for el in elements_with_id[:20]:
        print(f"  • Tag: <{el.name}> | ID: '{el.get('id')}'")

except requests.exceptions.RequestException as e:
    print(f"An error occurred: {e}")
```
