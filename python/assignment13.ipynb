<p>Create 'society_recursion.ipynb' for a fictional housing society. Write a recursive function `sum_dues(dues_list)` that totals 12 monthly maintenance dues you type, with a base case for an empty list and `dues_list[1:]` in the recursive call. Print its result next to `sum(dues_list)` for comparison.</p>

```python
# 12 monthly maintenance dues
dues_list = [1500, 1500, 1600, 1600, 1700, 1700,
             1800, 1800, 1900, 1900, 2000, 2000]

# Recursive function
def sum_dues(dues_list):
    if len(dues_list) == 0:  # Base case
        return 0
    return dues_list[0] + sum_dues(dues_list[1:])

# Compare results
print("Total using recursion:", sum_dues(dues_list))
print("Total using sum():", sum(dues_list))
```

<p>Write a recursive function `total_nested(block)` that totals dues stored in a nested list: 3 floors, each holding 4 flat dues. Use `type(item) is list` to decide when to recurse, print the grand total, and check it equals the sum of all 12 values.</p>

```python
# 3 floors, each with 4 flat dues
block = [
    [1500, 1600, 1700, 1800],
    [1500, 1600, 1700, 1800],
    [1500, 1600, 1700, 1800]
]

def total_nested(block):
    total = 0
    for item in block:
        if type(item) is list:
            total += total_nested(item)
        else:
            total += item
    return total

# Calculate and verify
grand_total = total_nested(block)
check_total = sum(sum(floor) for floor in block)

print("Grand Total:", grand_total)
print("Sum of all 12 dues:", check_total)
print("Are both equal?", grand_total == check_total)
```

<p>Write a recursive function `late_fee_amount(amount, months, depth=0)` that grows the due by 2 percent per month and prints one line per call with indentation based on depth. Call it with Rs 2,000 for 4 months, then compare the answer with a `for` loop version. Hint: A very large `months` will hit the recursion limit.</p>

```python
# Recursive function: 2% late fee per month
def late_fee_amount(amount, months, depth=0):
    print("  " * depth, "Amount:", round(amount, 2))

    if months == 0:
        return amount

    return late_fee_amount(amount * 1.02, months - 1, depth + 1)


# Using recursion
recursive_total = late_fee_amount(2000, 4)
print("Total using recursion: Rs.", round(recursive_total, 2))


# Using for loop
amount = 2000

for i in range(4):
    amount = amount * 1.02

print("Total using for loop: Rs.", round(amount, 2))
```
