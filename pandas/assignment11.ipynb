Use pd.concat to combine two DataFrames representing January and February orders from a Swiggy-style food delivery app, stacking them row-wise to create a single DataFrame with all orders.

```python
jan_orders = pd.DataFrame({
    'OrderID': ['SWG-101', 'SWG-102', 'SWG-103'],
    'Restaurant': ["La Pino's Pizza", "McDonald's", "Subway"],
    'Amount': [450, 299, 180]
})

feb_orders = pd.DataFrame({
    'OrderID': ['SWG-104', 'SWG-105', 'SWG-106'],
    'Restaurant': ["Domino's", "Haldiram's", "KFC"],
    'Amount': [520, 310, 420]
})

all_orders = pd.concat([jan_orders, feb_orders], ignore_index=True)
all_orders
```

Given two DataFrames — one with Flipkart product details and another with their respective ratings — use pd.concat to merge them column-wise so each product row includes its rating.

Create a DataFrame showing daily song streams for three users on Spotify for one week, then use the pivot function to reshape the data so each row is a user and each column is a day.

```python
data = {
    'User': ['Alice', 'Alice', 'Alice', 'Alice', 'Alice', 'Alice', 'Alice',
             'Bob', 'Bob', 'Bob', 'Bob', 'Bob', 'Bob', 'Bob',
             'Charlie', 'Charlie', 'Charlie', 'Charlie', 'Charlie', 'Charlie', 'Charlie'],
    'Day': ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun',
            'Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun',
            'Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'],
    'Streams': [12, 18, 15, 22, 35, 40, 28,
                8, 10, 14, 12, 25, 30, 20,
                5, 12, 9, 15, 18, 22, 16]
}

df = pd.DataFrame(data)

# reshape:
df_pivot = df.pivot(index='User', columns='Day', values='Streams')

# reorder:
days_order = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
df_pivot = df_pivot[days_order]

df_pivot
```

Given a DataFrame of Zomato restaurant orders (with columns: restaurant, month, total_amount), use pivot_table to calculate the average order amount per restaurant for each month.

```python
data = {
    'restaurant': ["La Pino's Pizza", "La Pino's Pizza", "McDonald's", "McDonald's", "Subway", "Subway", "La Pino's Pizza", "McDonald's"],
    'month': ['Jan', 'Jan', 'Jan', 'Feb', 'Jan', 'Feb', 'Feb', 'Feb'],
    'total_amount': [450, 350, 250, 300, 180, 220, 500, 350]
}

df = pd.DataFrame(data)

df_pivot = df.pivot_table(index='restaurant', columns='month', values='total_amount', aggfunc='mean')
df_pivot
```
