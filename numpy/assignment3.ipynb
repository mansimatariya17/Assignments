Create a NumPy array representing the number of likes on 7 Instagram posts and print its ndim, shape, size, dtype, itemsize, and nbytes properties.

```python
import numpy as np
```

```python
instagram_likes = np.array([450,780,620,970,832,1000,745])

print('instagram likes:',instagram_likes)
print('number of dimensions:', instagram_likes.ndim)
print('Shape:', instagram_likes.shape)
print('size:', instagram_likes.size)
print('Data Type:', instagram_likes.dtype)
print('Item-size:', instagram_likes.itemsize)
print('nbytes:', instagram_likes.nbytes)
```

Given a 2D NumPy array of daily step counts for 5 days (each row is a day, columns are morning and evening), use reshape() to convert it into a 1D array, then back to a 2D array with 5 rows and 2 columns.<br><br><em><strong>Hint:</strong> Use the shape attribute to check your array after each reshape.</em>

```python
step_counts = np.array([[120,450,330,156,745],[478,126,451,956,785]])

print('1D array:',step_counts.reshape(-1))
print('2D array:',step_counts.reshape(5,2))
```


Build a NumPy array representing the prices of 12 food items from a Zomato order, then use ravel(), flatten(), and resize() to create different shaped versions of the data and print each result.<br><br><em><strong>Constraint:</strong> Show the difference between ravel() and flatten() in your code comments.</em>

```python
arr = np.array([[50,60,150,80],[65,45,63,52],[41,71,56,96]])
print(arr)
a = arr

#ravel(If we change value in this array it will be changed in original matrix to)
a1 = a.ravel()
print('Ravel:',a1)

#flatten(If we change value in this array it will not be changed in original matrix.)
a2 = a.flatten()
print('Flatten:',a2)

#resize
a3 = np.resize(arr, (2,2))
print('Resize:',a3)
```


Take a 3x3 NumPy array representing a mini Spotify playlist grid (rows: playlists, columns: song counts in categories like Pop, Rock, Indie). Use both T and np.transpose() to swap rows and columns, then print the transposed array.

```python
spotify_palylist = np.array([['Pop','Rock','Indie'],[10,15,25],[30,12,17]])
print(spotify_palylist)

#Transpose
Transpose = np.transpose(spotify_palylist)
print(Transpose)

#T
T = spotify_palylist.T
print(T)
```
