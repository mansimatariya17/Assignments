```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Load a CSV file containing Flipkart product prices and use sns.boxplot() to visualize the price distribution.

```python
df = pd.DataFrame({
    'Product': ['Phone', 'Laptop', 'Headphones', 'Watch', 'Tablet',
                'Camera', 'Speaker', 'Keyboard', 'Monitor', 'Mouse'],
    'Price': [15000, 55000, 1200, 2500, 18000,
              35000, 3000, 1500, 12000, 500]
})

# Create a boxplot of product prices
sns.boxplot(x=df['Price'], color="#CA5718")

plt.title('Distribution of Flipkart Product Prices')
plt.xlabel('Price (₹)')
plt.show()
```

Use sns.violinplot() to compare the order amounts for two different Zomato restaurant categories (e.g., 'Fast Food' vs 'Desserts') from a sample dataset.

```python
df = pd.DataFrame({
    'Category': [
        'Fast Food', 'Fast Food', 'Fast Food', 'Fast Food', 'Fast Food',
        'Desserts', 'Desserts', 'Desserts', 'Desserts', 'Desserts'
    ],
    'Order_Amount': [
        150, 250, 300, 200, 350,
        100, 180, 220, 150, 280
    ]
})

# Create a violin plot
sns.violinplot(data=df, x='Category', y='Order_Amount')

plt.title('Order Amount Comparison: Fast Food vs Desserts')
plt.xlabel('Restaurant Category')
plt.ylabel('Order Amount (₹)')
plt.show()
```

Given an array of IPL match ticket prices, plot a boxplot using sns.boxplot() and identify any outliers in the data.<br><br><em><strong>Hint:</strong> Outliers are points that appear outside the whiskers of the boxplot.</em>

```python
ticket_prices = [
    500, 700, 800, 1000, 1200, 1500, 1800,
    2000, 2200, 2500, 3000, 3500, 4000, 15000
]

# Create a boxplot
plt.figure(figsize=(10,6))
sns.boxplot(x=ticket_prices)

plt.title('Distribution of IPL Match Ticket Prices')
plt.xlabel('Ticket Price (₹)')
plt.show()
```

Create a violin plot using sns.violinplot() to show the distribution of daily step counts collected from a fitness app like Google Fit, and explain what the shape of the plot tells you about user activity.

```python
steps = [
    3000, 4500, 5000, 6000, 7500, 8000, 6500, 9000, 10000, 5500,
    4000, 7000, 8500, 9500, 11000, 6000, 7500, 5000, 12000, 10500,
    3500, 6500, 8000, 9000, 7000, 5500, 13000, 10000, 4500, 8500
]

df = pd.DataFrame({'Daily_Steps': steps})

# Create violin plot
plt.figure(figsize=(9,6))
sns.violinplot(y=df['Daily_Steps'])

plt.title('Distribution of Daily Step Counts')
plt.ylabel('Number of Steps')
plt.show()
```
