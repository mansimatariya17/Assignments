<p>Create 'upi_files.ipynb' for fictional UPI-style payments. With `open()` in 'w' mode, write 'upi_log.txt' with 10 lines in the format txn_id,date,payee,amount_inr,status (dates as DD/MM/YYYY, invented payees, status SUCCESS or FAILED). Append 2 more lines in 'a' mode, read it back with `with open(...)` and print the line count. Hint: Add '\n' to every line and pass `encoding='utf-8'`.</p>

```python
# Write 10 transactions in write mode
transactions = [
    "TXN001,01/11/2026,Rahul,500,SUCCESS\n",
    "TXN002,02/11/2026,Priya,1200,SUCCESS\n",
    "TXN003,03/11/2026,Amit,750,FAILED\n",
    "TXN004,04/11/2026,Neha,2000,SUCCESS\n",
    "TXN005,05/11/2026,Karan,450,SUCCESS\n",
    "TXN006,06/11/2026,Riya,900,FAILED\n",
    "TXN007,07/11/2026,Jay,1500,SUCCESS\n",
    "TXN008,08/11/2026,Pooja,600,SUCCESS\n",
    "TXN009,09/11/2026,Ankit,1100,SUCCESS\n",
    "TXN010,10/11/2026,Meera,800,FAILED\n"
]

with open("upi_log.txt", "w", encoding="utf-8") as file:
    file.writelines(transactions)

# Append 2 more transactions
with open("upi_log.txt", "a", encoding="utf-8") as file:
    file.write("TXN011,11/11/2026,Arjun,2500,SUCCESS\n")
    file.write("TXN012,12/11/2026,Kavya,700,SUCCESS\n")

# Read file and count lines
with open("upi_log.txt", "r", encoding="utf-8") as file:
    data = file.readlines()

print("UPI Transaction Log:")
print("".join(data))
print("Total number of transactions:", len(data))
```

<p>Create 'upi_payments.csv' with a header and 12 rows using `csv.writer`, then read it with `csv.DictReader` and print the total of amount_inr for SUCCESS rows, converting the amounts with `float()`. Hint: Open the file with `newline=''` when writing CSV.</p>

```python
import csv

# Create CSV file with header and 12 transactions
transactions = [
    ["TXN001", "01/11/2026", "Rahul", 500, "SUCCESS"],
    ["TXN002", "02/11/2026", "Priya", 1200, "SUCCESS"],
    ["TXN003", "03/11/2026", "Amit", 750, "FAILED"],
    ["TXN004", "04/11/2026", "Neha", 2000, "SUCCESS"],
    ["TXN005", "05/11/2026", "Karan", 450, "SUCCESS"],
    ["TXN006", "06/11/2026", "Riya", 900, "FAILED"],
    ["TXN007", "07/11/2026", "Jay", 1500, "SUCCESS"],
    ["TXN008", "08/11/2026", "Pooja", 600, "SUCCESS"],
    ["TXN009", "09/11/2026", "Ankit", 1100, "SUCCESS"],
    ["TXN010", "10/11/2026", "Meera", 800, "FAILED"],
    ["TXN011", "11/11/2026", "Arjun", 2500, "SUCCESS"],
    ["TXN012", "12/11/2026", "Kavya", 700, "SUCCESS"]
]

# Write CSV file
with open("upi_payments.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow(["txn_id", "date", "payee", "amount_inr", "status"])
    writer.writerows(transactions)

# Read CSV and calculate successful payments
total = 0

with open("upi_payments.csv", "r", encoding="utf-8") as file:
    reader = csv.DictReader(file)

    for row in reader:
        if row["status"] == "SUCCESS":
            total += float(row["amount_inr"])

print("Total SUCCESS payments: Rs.", total)
```

<p>Build the dictionary payee_summary with each payee mapped to the total of its SUCCESS amounts, save it as 'payee_summary.json' using `json.dump(..., indent=2)`, load it back with `json.load()` and print the payee with the highest total.</p>

```python
import csv
import json

# Build payee summary
payee_summary = {}

with open("upi_payments.csv", "r", encoding="utf-8") as file:
    reader = csv.DictReader(file)

    for row in reader:
        if row["status"] == "SUCCESS":
            payee = row["payee"]
            amount = float(row["amount_inr"])

            payee_summary[payee] = payee_summary.get(payee, 0) + amount

# Save dictionary to JSON
with open("payee_summary.json", "w", encoding="utf-8") as file:
    json.dump(payee_summary, file, indent=2)

# Load JSON file
with open("payee_summary.json", "r", encoding="utf-8") as file:
    loaded_summary = json.load(file)

# Find payee with highest total
highest_payee = max(loaded_summary, key=loaded_summary.get)

print("Payee Summary:", loaded_summary)
print("Highest Payee:", highest_payee)
print("Highest Total: Rs.", loaded_summary[highest_payee])
```

<p>Finish with `pathlib`: create the folder 'upi_output' using `Path('upi_output').mkdir(exist_ok=True)`, move the three output files into it by re-saving them there, check each with `.exists()`, print each file's size from `.stat().st_size`, and list the CSV files with `glob('*.csv')`. Run Restart & Run All.</p>

```python
from pathlib import Path
import shutil

# Create output folder
output_dir = Path("upi_output")
output_dir.mkdir(exist_ok=True)

# Move/re-save three output files into the folder
files = ["upi_log.txt", "upi_payments.csv", "payee_summary.json"]

for filename in files:
    source = Path(filename)
    destination = output_dir / filename
    shutil.copy(source, destination)

# Check existence and print file sizes
for filename in files:
    file_path = output_dir / filename

    print(filename)
    print("Exists:", file_path.exists())
    print("Size:", file_path.stat().st_size, "bytes")
    print()

# List CSV files
print("CSV files:", list(output_dir.glob("*.csv")))
```
