Use the requests library in Python to make a GET request to https://jsonplaceholder.typicode.com/posts and print the status code and first 100 characters of the response text.


```python
import requests

# Define the endpoint URL
url = "https://jsonplaceholder.typicode.com/posts"

try:
    # Make a GET request to the API
    response = requests.get(url)
    
    # Print the HTTP status code
    print(f"Status Code: {response.status_code}")
    
    # Print the first 100 characters of the response body text
    print("\nFirst 100 characters of response text:")
    print(response.text[:100])

except requests.exceptions.RequestException as e:
    print(f"An error occurred: {e}")
```

Send a GET request to https://httpbin.org/headers with a custom User-Agent header set to 'FlipkartBot/1.0' and print the User-Agent value received by the server from the response.

```python
import requests

# Define the endpoint and custom headers
url = "https://httpbin.org/headers"
headers = {
    "User-Agent": "FlipkartBot/1.0"
}

try:
    # Send GET request with custom headers
    response = requests.get(url, headers=headers)
    response.raise_for_status()
    
    # Parse the JSON response
    response_data = response.json()
    
    # Extract and print the User-Agent value received by the server
    received_user_agent = response_data["headers"].get("User-Agent")
    print(f"User-Agent received by server: {received_user_agent}")

except requests.exceptions.RequestException as e:
    print(f"An error occurred: {e}")
```

Simulate logging into a Zomato-style food delivery site by sending a POST request to https://httpbin.org/post with form data {'username': 'foodie', 'password': 'tasty123'} using requests. Print the JSON response to confirm your data was sent.<br><br><em><strong>Hint:</strong> Use the data parameter in requests.post().</em>

```python
import requests

# Define the endpoint URL
url = "https://httpbin.org/post"

# Define the form data payload
form_data = {
    "username": "foodie",
    "password": "tasty123"
}

try:
    # Send a POST request using the 'data' parameter for form data
    response = requests.post(url, data=form_data)
    
    # Check for HTTP errors
    response.raise_for_status()
    
    # Print the status code and the JSON response returned by the server
    print(f"Status Code: {response.status_code}\n")
    print("JSON Response received from server:")
    print(response.json())

except requests.exceptions.RequestException as e:
    print(f"An error occurred: {e}")
```

Make a GET request to https://httpbin.org/cookies/set?sessionid=12345 using requests, then make another GET request to https://httpbin.org/cookies to check if your sessionid cookie is being sent. Print the cookies received in the second response.<br><br><em><strong>Hint:</strong> Use requests.Session() to persist cookies between requests.</em>

```python
import requests

# Create a session object to automatically store and send cookies
with requests.Session() as session:
    
    # Step 1: Make a GET request to set the cookie
    set_url = "https://httpbin.org/cookies/set?sessionid=12345"
    response_set = session.get(set_url)
    print(f"Set Cookie Status Code: {response_set.status_code}")
    
    # Step 2: Make a second GET request to check if the sessionid cookie is sent
    get_url = "https://httpbin.org/cookies"
    response_get = session.get(get_url)
    print(f"Check Cookies Status Code: {response_get.status_code}\n")
    
    # Print the JSON response containing the cookies received by the server
    print("JSON Response from /cookies:")
    print(response_get.json())
```

Write a function check_status_code(url) that makes a GET request to the given URL and prints a message: 'Success' if status code is 200, 'Not Found' if 404, and 'Other Error' for anything else.<br><br><em><strong>Constraint:</strong> Use if-elif-else to handle the different status codes.</em>

```python
import requests

def check_status_code(url):
    try:
        # Make a GET request to the URL
        response = requests.get(url)
        
        # Check the status code using if-elif-else
        if response.status_code == 200:
            print("Success")
        elif response.status_code == 404:
            print("Not Found")
        else:
            print("Other Error")
            
    except requests.exceptions.RequestException as e:
        print("Other Error")
```
