
Create a DataFrame named 'movies' with columns: 'title', 'rating', and 'release_year' for at least 6 popular Bollywood films. Use sort_values() to display the movies sorted by 'rating' from highest to lowest.

```python
movies = pd.DataFrame({
    'title': [
        '3 Idiots',
        'Dangal',
        'Sholay',
        'Zindagi Na Milegi Dobara',
        'Taare Zameen Par',
        'Lagaan'
    ],
    'rating': [8.4, 8.3, 8.1, 8.2, 8.3, 8.1],
    'release_year': [2009, 2016, 1975, 2011, 2007, 2001]
})

result = movies.sort_values(by = 'rating',ascending=False)
print(result)
```

Given a DataFrame of IPL teams with columns 'team', 'points', and 'net_run_rate', use sort_index() to sort the DataFrame by the default index in descending order, then print the result.<br><br><em><strong>Hint:</strong> Use the ascending parameter in sort_index().</em>

```python
teams = pd.DataFrame({
    'team': ['CSK', 'MI', 'RCB', 'KKR', 'GT', 'RR'],
    'points': [16, 14, 18, 16, 12, 14],
    'net_run_rate': [0.25, 0.15, 0.45, 0.30, -0.10, 0.20]
})

result = teams.sort_index(ascending=False)
print(result)
```

Reset the index of a DataFrame containing a list of trending YouTube videos (columns: 'video_title', 'views', 'likes'), and make sure the old index is not added as a column in the result.<br><br><em><strong>Constraint:</strong> Use the drop=True argument in reset_index().</em>

```python
videos = pd.DataFrame({
    'video_title': ['Music Video', 'Funny Shorts', 'Vlog', 'Dance Reel'],
    'views': [1000000, 500000, 750000, 900000],
    'likes': [50000, 25000, 35000, 45000]
})

result = videos.reset_index(drop=True)
print(result)
```


You have a DataFrame of Zomato restaurant names and their ratings. Use reindex() to rearrange the rows so that your favorite restaurant appears first, followed by the rest in any order.

```python
zomato = pd.DataFrame({
    'restaurant': ['McDonalds', 'Dominos', 'Subway', 'Pizza Hut'],
    'rating': [4.2, 4.5, 4.0, 4.3]
})

result = zomato.reindex([1,0,2,3])
print(result)
```

Take a DataFrame of Flipkart products with columns 'product_name', 'price', and 'discount'. First, sort the products by 'discount' descending, then reset the index so it starts from 0 and is sequential after sorting.

```python
products = pd.DataFrame({
    'product_name': ['Phone', 'Laptop', 'Headphones', 'Smartwatch', 'Tablet'],
    'price': [15000, 55000, 2000, 3000, 20000],
    'discount': [10, 25, 15, 30, 20]
})

result = products.sort_values(by = 'discount',ascending=False).reset_index()
print(result)
```
