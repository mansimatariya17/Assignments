<p>Add the method `resale_estimate(self)` to `Vehicle` that returns `price_inr * (1 - 0.08) ** age` with age as 2026 minus reg_year, and override it in `Bike` (10 percent a year) and `Scooter` (7 percent a year). Print rounded estimates for one Bike and one Scooter.</p>

```python
class Vehicle:
    def __init__(self, model_name: str, reg_year: int, price_inr: float):
        self.model_name = model_name
        self.reg_year = reg_year
        self.price_inr = price_inr

    def resale_estimate(self) -> float:
        age = 2026 - self.reg_year
        return self.price_inr * (1 - 0.08) ** age


class Bike(Vehicle):
    def __init__(self, model_name: str, reg_year: int, price_inr: float, engine_cc: int):
        super().__init__(model_name, reg_year, price_inr)
        self.engine_cc = engine_cc

    def resale_estimate(self) -> float:
        age = 2026 - self.reg_year
        return self.price_inr * (1 - 0.10) ** age


class Scooter(Vehicle):
    def __init__(self, model_name: str, reg_year: int, price_inr: float, boot_litres: float):
        super().__init__(model_name, reg_year, price_inr)
        self.boot_litres = boot_litres

    def resale_estimate(self) -> float:
        age = 2026 - self.reg_year
        return self.price_inr * (1 - 0.07) ** age


bike = Bike("Roadster 160", 2023, 85000, 160)
scooter = Scooter("CityGlide 125", 2024, 72000, 18.5)

print("Bike attributes:", vars(bike))
print("Scooter attributes:", vars(scooter))
print(f"{bike.model_name} resale estimate: ₹{round(bike.resale_estimate(), 2)}")
print(f"{scooter.model_name} resale estimate: ₹{round(scooter.resale_estimate(), 2)}")
```

<p>Create a list of 6 mixed Bike and Scooter objects and loop through it calling `resale_estimate()` with one identical line of code for all of them, printing the type name, model and estimate. Print the total of all estimates.</p>

```python
class Vehicle:
    def __init__(self, model_name: str, reg_year: int, price_inr: float):
        self.model_name = model_name
        self.reg_year = reg_year
        self.price_inr = price_inr

    def resale_estimate(self) -> float:
        age = 2026 - self.reg_year
        return self.price_inr * (1 - 0.08) ** age


class Bike(Vehicle):
    def __init__(self, model_name: str, reg_year: int, price_inr: float, engine_cc: int):
        super().__init__(model_name, reg_year, price_inr)
        self.engine_cc = engine_cc

    def resale_estimate(self) -> float:
        age = 2026 - self.reg_year
        return self.price_inr * (1 - 0.10) ** age


class Scooter(Vehicle):
    def __init__(self, model_name: str, reg_year: int, price_inr: float, boot_litres: float):
        super().__init__(model_name, reg_year, price_inr)
        self.boot_litres = boot_litres

    def resale_estimate(self) -> float:
        age = 2026 - self.reg_year
        return self.price_inr * (1 - 0.07) ** age


vehicles = [
    Bike("Roadster 160", 2023, 85000, 160),
    Scooter("CityGlide 125", 2024, 72000, 18.5),
    Bike("Cruiser 350", 2022, 165000, 349),
    Scooter("UrbanZip 110", 2025, 68000, 16.0),
    Bike("StreetPulse 200", 2024, 120000, 199),
    Scooter("MetroLite 125", 2021, 78000, 21.0),
]

total_estimate = 0.0

for vehicle in vehicles:
    estimate = vehicle.resale_estimate()
    total_estimate += estimate
    print(f"{type(vehicle).__name__} | {vehicle.model_name} | ₹{round(estimate, 2)}")

print(f"Total resale estimate across {len(vehicles)} vehicles: ₹{round(total_estimate, 2)}")
```

<p>Finish by sorting the 6 objects with `sorted()` and a lambda on `resale_estimate()` in descending order, saving 'two_wheeler_estimates.csv' with the header type,model,age_years,resale_estimate_inr using `csv.writer`, and reading it back to confirm 6 data rows. Run Restart & Run All.</p>

```python
import csv


class Vehicle:
    def __init__(self, model_name: str, reg_year: int, price_inr: float):
        self.model_name = model_name
        self.reg_year = reg_year
        self.price_inr = price_inr

    def resale_estimate(self) -> float:
        age = 2026 - self.reg_year
        return self.price_inr * (1 - 0.08) ** age


class Bike(Vehicle):
    def __init__(self, model_name: str, reg_year: int, price_inr: float, engine_cc: int):
        super().__init__(model_name, reg_year, price_inr)
        self.engine_cc = engine_cc

    def resale_estimate(self) -> float:
        age = 2026 - self.reg_year
        return self.price_inr * (1 - 0.10) ** age


class Scooter(Vehicle):
    def __init__(self, model_name: str, reg_year: int, price_inr: float, boot_litres: float):
        super().__init__(model_name, reg_year, price_inr)
        self.boot_litres = boot_litres

    def resale_estimate(self) -> float:
        age = 2026 - self.reg_year
        return self.price_inr * (1 - 0.07) ** age


vehicles = [
    Bike("Roadster 160", 2023, 85000, 160),
    Scooter("CityGlide 125", 2024, 72000, 18.5),
    Bike("Cruiser 350", 2022, 165000, 349),
    Scooter("UrbanZip 110", 2025, 68000, 16.0),
    Bike("StreetPulse 200", 2024, 120000, 199),
    Scooter("MetroLite 125", 2021, 78000, 21.0),
]

total_estimate = 0.0
for vehicle in vehicles:
    estimate = vehicle.resale_estimate()
    total_estimate += estimate
    print(f"{type(vehicle).__name__} | {vehicle.model_name} | ₹{round(estimate, 2)}")

print(f"Total resale estimate across {len(vehicles)} vehicles: ₹{round(total_estimate, 2)}")

# Sort descending by resale_estimate() using sorted() and a lambda
sorted_vehicles = sorted(vehicles, key=lambda v: v.resale_estimate(), reverse=True)

# Write CSV using csv.writer
csv_filename = "two_wheeler_estimates.csv"
with open(csv_filename, mode="w", newline="", encoding="utf-8") as csv_file:
    writer = csv.writer(csv_file)
    writer.writerow(["type", "model", "age_years", "resale_estimate_inr"])
    for v in sorted_vehicles:
        age_years = 2026 - v.reg_year
        writer.writerow([type(v).__name__, v.model_name, age_years, round(v.resale_estimate(), 2)])

# Read back and verify 6 data rows
with open(csv_filename, mode="r", newline="", encoding="utf-8") as read_file:
    reader = csv.reader(read_file)
    rows = list(reader)

header = rows[0]
data_rows = rows[1:]

print(f"\nHeader: {header}")
print(f"Confirmed data rows: {len(data_rows)}")
for row in data_rows:
    print(row)
```
