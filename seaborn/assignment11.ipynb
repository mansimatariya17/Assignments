```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Create a Seaborn barplot showing the number of orders per food category (e.g., Pizza, Burger, Biryani, Momos, Ice Cream) and save the chart as a PNG file with 300 DPI.

```python
df = pd.DataFrame({
    'food_category': ['Pizza', 'Burger', 'Biryani', 'Momos', 'Ice Cream'],
    'number_of_orders': [120, 95, 150, 80, 110]
})


# Create barplot
plt.figure(figsize=(9, 6))

sns.barplot(
    data=df,
    x='food_category',
    y='number_of_orders',
    hue='food_category',
    palette='deep',
    legend=False
)

plt.title('Number of Orders by Food Category')
plt.xlabel('Food Category')
plt.ylabel('Number of Orders')

plt.tight_layout()

# Save chart as PNG with 300 DPI
plt.savefig('food_orders_by_category.png', dpi=300, bbox_inches='tight')

# Display chart
plt.show()
```

Generate a Seaborn line plot visualizing daily step counts for one week, then export the chart as an SVG file suitable for use in a PowerPoint presentation.<br><br><em><strong>Hint:</strong> Use plt.savefig('filename.svg', format='svg').</em>

```python
df = pd.DataFrame({
    'day': ['Monday', 'Tuesday', 'Wednesday', 'Thursday',
            'Friday', 'Saturday', 'Sunday'],
    'steps': [4500, 6200, 5800, 7500, 8100, 9200, 7800]
})

# Apply Seaborn theme
sns.set_theme(style='whitegrid', palette='deep', context='talk')

# Create line plot
plt.figure(figsize=(10, 5))

sns.lineplot(
    data=df,
    x='day',
    y='steps',
    marker='o',
    linewidth=2.5
)

# Add title and labels
plt.title('Daily Step Counts Over One Week')
plt.xlabel('Day')
plt.ylabel('Number of Steps')

plt.xticks(rotation=20)
plt.tight_layout()

# Save chart as SVG
plt.savefig('weekly_step_counts.svg', format='svg', bbox_inches='tight')

# Display chart
plt.show()
```

Customize a Seaborn chart using the 'darkgrid' style and a color palette of your choice, then save it as a high-resolution PNG image with 1200 DPI.

```python
df = pd.DataFrame({
    'product': ['Mobile', 'Laptop', 'Tablet', 'Headphones', 'Smartwatch'],
    'sales': [150, 120, 90, 180, 110]
})

# Apply Seaborn style and color palette
sns.set_theme(
    style='darkgrid',
    palette='colorblind',
    context='notebook'
)

# Create barplot
plt.figure(figsize=(10, 6))

sns.barplot(
    data=df,
    x='product',
    y='sales',
    hue='product',
    legend=False
)

# Add title and labels
plt.title('Product Sales Dashboard', fontsize=16)
plt.xlabel('Product Category')
plt.ylabel('Number of Sales')

plt.tight_layout()

# Save as high-resolution PNG (1200 DPI)
plt.savefig(
    'product_sales_1200dpi.png',
    format='png',
    dpi=1200,
    bbox_inches='tight'
)

# Display chart
plt.show()
```

Create a template function called apply_presentation_theme() that applies a consistent Seaborn style, font, and color palette to any chart, then use it to style and export a pie chart showing your top 5 most-used apps.

```python
def apply_presentation_theme():
    sns.set_theme(
        style="darkgrid",
        palette="colorblind",
        font="Arial",
        font_scale=1.2
    )

# Apply the theme
apply_presentation_theme()

# Top 5 most-used apps (sample data)
apps = ['Instagram', 'WhatsApp', 'YouTube', 'Facebook', 'Spotify']
usage = [30, 25, 20, 15, 10]

# Create pie chart
plt.figure(figsize=(8, 6))

plt.pie(
    usage,
    labels=apps,
    autopct='%1.1f%%',
    startangle=90,
    colors=sns.color_palette("colorblind", 5),
    explode=(0.05, 0, 0, 0, 0)
)

plt.title("Top 5 Most-Used Apps", fontsize=16)
plt.axis('equal')

# Export chart
plt.savefig("top_5_apps.png", dpi=300, bbox_inches='tight')

# Display chart
plt.show()
```

Use ChatGPT to generate a code snippet for exporting a Seaborn heatmap as a 1200 DPI SVG file, then test and adapt the code to save your own chart.<br><br><em><strong>Hint:</strong> Paste your prompt and the code you received as a comment in your script.</em>

```python
# Sample dataset
data = {
    'Math': [85, 90, 78, 92, 88],
    'Science': [80, 85, 88, 90, 82],
    'English': [75, 80, 85, 88, 90],
    'Computer': [95, 92, 90, 98, 96]
}

df = pd.DataFrame(data)

# Calculate correlation matrix
correlation = df.corr()

# Apply Seaborn theme
sns.set_theme(style="whitegrid")

# Create heatmap
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    linewidths=0.5,
    fmt=".2f"
)

plt.title("Student Subject Correlation Heatmap")

# Export as SVG with 1200 DPI
plt.savefig(
    "student_correlation_heatmap.svg",
    format="svg",
    dpi=1200,
    bbox_inches="tight"
)

# Display chart
plt.show()

print("Heatmap successfully exported as SVG!")
```
