Install Matplotlib in your Python environment and write a script named hello_plot.py that creates a simple line plot showing the number of daily steps you walked over the past 7 days.

```python
steps = [550,400,500,659,700,487,900]
days = ['Day 1','Day 2','Day 3','Day 4','Day 5','Day 6','Day 7']
plt.figure(figsize=(8,5))
plt.bar(days, steps, color='purple')
plt.title('Daily Steps Over the Past 7 Days')
plt.xlabel('Days')
plt.ylabel('Steps')
plt.tight_layout()
plt.show()
```

Write a Python script that creates a Matplotlib figure with two Axes: one showing your daily Spotify listening hours for a week and the other showing your daily YouTube viewing hours for the same week. Display both plots in a single figure.


```python
week = ['Mon','Tue','Wed','Thu','Fri','Sat','Sun']

spotify_listening_hours = [2,4,5,8,3,7,9]
youtube_viewing_hours = [4,3,9,6,8,7,3]

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.bar(week, spotify_listening_hours, color='#6C63FF')
plt.title('spotify_listening_hours')
plt.xlabel('week')
plt.ylabel('listening_hours')

plt.subplot(1,2,2)
plt.bar(week, youtube_viewing_hours, color="#2C9380")
plt.title('youtube_viewing_hours')
plt.xlabel('week')
plt.ylabel('viewing_hours')

plt.tight_layout()
plt.show()
```
