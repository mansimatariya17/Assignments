```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Import Seaborn and use sns.histplot() to visualize the distribution of delivery times (in minutes) from a sample Swiggy order dataset with at least 30 values.

```python
delivery_time = [
    25, 30, 35, 40, 28, 32, 45, 38, 22, 27,
    33, 36, 42, 29, 31, 24, 39, 34, 26, 41,
    37, 30, 28, 44, 35, 32, 23, 40, 33, 29
]

sns.histplot(delivery_time, bins=8, kde=True)

plt.title('Distribution of Swiggy Delivery Times')
plt.xlabel('Delivery Time (minutes)')
plt.ylabel('Frequency')
plt.show()
```

Download or create a list of daily step counts (e.g., from a fitness app) for one month and plot the distribution using sns.kdeplot(), showing the probability density curve.

```python
steps = [
    4500, 6200, 7800, 5500, 8900, 10200, 7100, 6800, 9500, 8300,
    4200, 7600, 11000, 9800, 6400, 8700, 7200, 5600, 9200, 10500,
    8000, 6900, 11500, 7400, 8800, 6000, 9700, 8200, 5300, 10000
]

# Plot probability density curve
sns.kdeplot(x=steps, fill=True)

plt.title('Distribution of Daily Step Counts (30 Days)')
plt.xlabel('Daily Step Count')
plt.ylabel('Probability Density')
plt.show()
```

Combine sns.histplot() and sns.kdeplot() to visualize the distribution of monthly spending amounts (in rupees) on Zomato orders, using the same dataset for both plots on a single figure.<br><br><em><strong>Hint:</strong> Use the same data and overlay the KDE curve on the histogram by calling both functions on the same axes.</em>

```python
spending = [
    150, 250, 300, 450, 200, 500, 350, 600, 700, 400,
    250, 800, 550, 300, 450, 900, 650, 350, 500, 750,
    200, 400, 600, 850, 300, 550, 700, 450, 1000, 650
]

# Plot histogram and KDE curve on the same figure
sns.histplot(x=spending, bins=10, kde=True, stat='density')

sns.kdeplot(x=spending, color="#E56741", linewidth=2)

plt.title('Distribution of Monthly Zomato Spending')
plt.xlabel('Monthly Spending (₹)')
plt.ylabel('Probability Density')
plt.show()
```

Given a dataset of IPL match scores, use sns.histplot() to plot the run distribution. Change the number of bins and observe how the plot changes. Write 2-3 lines about what you notice.<br><br><em><strong>Hint:</strong> Try bins=5, bins=10, and bins=20.</em>

```python
scores = [
    120, 145, 160, 175, 180, 190, 155, 200, 210, 165,
    140, 185, 195, 220, 170, 150, 205, 230, 175, 160,
    190, 215, 145, 180, 200, 240, 155, 170, 185, 225
]

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# Histogram with 5 bins
sns.histplot(x=scores, bins=5,ax=axes[0])
axes[0].set_title('5 Bins')
axes[0].set_xlabel('Match Score')
axes[0].set_ylabel('Frequency')

# Histogram with 10 bins
sns.histplot(x=scores, bins=10,ax=axes[1])
axes[1].set_title('10 Bins')
axes[1].set_xlabel('Match Score')
axes[1].set_ylabel('Frequency')
# Histogram with 20 bins
sns.histplot(x=scores, bins=20,ax=axes[2])
axes[2].set_title('20 Bins')
axes[2].set_xlabel('Match Score')
axes[2].set_ylabel('Frequency')
plt.show()
```
