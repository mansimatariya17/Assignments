```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Create a Pandas DataFrame with at least 5 columns representing different features of food delivery orders (e.g., delivery_time, order_value, tip_amount, rating, delivery_distance) and 20 rows of random data, then calculate and display the correlation matrix using the .corr() function.

```python
np.random.seed(42)

data = {
    'delivery_time': np.random.randint(15, 60, 20),
    'order_value': np.random.randint(100, 1000, 20),
    'tip_amount': np.random.randint(10, 100, 20),
    'rating': np.random.randint(1, 6, 20),
    'delivery_distance': np.random.uniform(1, 15, 20).round(2)
}

# Create DataFrame
df = pd.DataFrame(data)

# Display the DataFrame
print("Food Delivery Orders:")
print(df)

# Calculate correlation matrix
correlation_matrix = df.corr()

# Display the correlation matrix
print("\nCorrelation Matrix:")
print(correlation_matrix.round(2))
```

Using Seaborn, plot a heatmap of the correlation matrix you generated for the food delivery data, and add annotations to show the correlation values inside each cell.

```python
np.random.seed(42)

data = {
    'delivery_time': np.random.randint(15, 60, 20),
    'order_value': np.random.randint(100, 1000, 20),
    'tip_amount': np.random.randint(10, 100, 20),
    'rating': np.random.randint(1, 6, 20),
    'delivery_distance': np.random.uniform(1, 15, 20).round(2)
}

df = pd.DataFrame(data)

# Calculate correlation matrix
corr_matrix = df.corr()

# Plot heatmap
plt.figure(figsize=(8, 6))

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Correlation Heatmap of Food Delivery Data')
plt.show()
```

Change the color palette of your Seaborn heatmap to a diverging palette (such as 'coolwarm' or 'RdYlGn') to better visualize positive and negative correlations, and explain which two features in your data have the strongest negative correlation.<br><br><em><strong>Hint:</strong> Use the cmap parameter in sns.heatmap().</em>

```python
np.random.seed(42)

data = {
    'delivery_time': np.random.randint(15, 60, 20),
    'order_value': np.random.randint(100, 1000, 20),
    'tip_amount': np.random.randint(10, 100, 20),
    'rating': np.random.randint(1, 6, 20),
    'delivery_distance': np.random.uniform(1, 15, 20).round(2)
}

df = pd.DataFrame(data)

# Calculate correlation matrix
corr_matrix = df.corr()

# Plot heatmap with a diverging palette
plt.figure(figsize=(8, 6))

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap='RdYlGn',
    fmt='.2f',
    linewidths=0.5,
    vmin=-1,
    vmax=1,
    center=0
)

plt.title('Correlation Heatmap - Food Delivery Data')
plt.show()

# Find the strongest negative correlation
corr_pairs = corr_matrix.where(
    np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
).stack()

strongest_negative = corr_pairs.idxmin()
print("Features:", strongest_negative)
print("Correlation:", round(corr_pairs.min(), 2))
```

Take the Flipkart mobile phones dataset (or any small CSV with product features you can find online), compute the correlation matrix for at least 4 numeric columns (e.g., price, rating, RAM, battery_capacity), and generate a Seaborn heatmap with annotations and a diverging color palette.

```python
data = {
    'price': [14999, 24999, 39999, 12999, 54999,
              18999, 32999, 7999, 44999, 27999],
    'rating': [4.2, 4.4, 4.5, 4.1, 4.6,
               4.3, 4.4, 3.9, 4.5, 4.3],
    'ram_capacity': [4, 8, 12, 4, 12, 6, 8, 3, 12, 8],
    'battery_capacity': [5000, 5000, 4800, 6000, 5000,
                         5000, 5500, 5000, 4700, 6000]
}

df = pd.DataFrame(data)

features = [
    "price",
    "rating",
    "ram_capacity",
    "battery_capacity"
]

# Convert selected columns to numeric values
for col in features:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Calculate the correlation matrix
corr_matrix = df[features].corr()

print("Correlation Matrix:")
print(corr_matrix.round(2))

# Plot the heatmap
plt.figure(figsize=(8, 6))

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5,
    vmin=-1,
    vmax=1,
    center=0
)

plt.title("Smartphone Features Correlation Heatmap")
plt.tight_layout()
plt.show()
```
