<p>Create 'gym_stream.ipynb' for a fictional gym. Make a list of 6 member IDs, create an iterator with `iter()` and call `next()` on it 6 times, then call `next(it, 'END')` once more and print the result. Hint: Without the default argument the seventh call raises StopIteration.</p>

```python
# List of 6 gym member IDs
member_ids = ["GYM101", "GYM102", "GYM103", "GYM104", "GYM105", "GYM106"]

# Create an iterator
it = iter(member_ids)

# Call next() 6 times
print(next(it))
print(next(it))
print(next(it))
print(next(it))
print(next(it))
print(next(it))

# Call next() one more time with default value
print(next(it, "END"))
```

<p>Create three lists of 6 members: names, plans (Monthly, Quarterly or Yearly) and fees in rupees. Use `zip()` with `enumerate(start=1)` to print lines like '1. Name - Plan - Rs Fee', then build a dictionary of name to fee with `dict(zip(...))`.</p>

```python
# Three lists of 6 gym members
names = ["Rahul", "Priya", "Amit", "Neha", "Karan", "Riya"]

plans = ["Monthly", "Quarterly", "Yearly",
         "Monthly", "Yearly", "Quarterly"]

fees = [1000, 2500, 8000, 1000, 8000, 2500]

# Use zip() with enumerate(start=1)
for number, (name, plan, fee) in zip(
    enumerate(names, start=1), zip(names, plans, fees)
):
    print(f"{number}. {name} - {plan} - Rs {fee}")

# Create dictionary of name to fee
member_fees = dict(zip(names, fees))

print("\nMember Fee Dictionary:")
print(member_fees)
```

<p>Write a generator function `daily_checkins(days)` that uses `yield` to return a check-in count of `40 + (day * 7) % 25` for each day from 1 to `days`. Print the first 3 values with `next()`, then loop the rest of a 14-day run, and print the total.</p>

```python
# Generator function
def daily_checkins(days):
    for day in range(1, days + 1):
        yield 40 + (day * 7) % 25

# Create generator for 14 days
checkins = daily_checkins(14)

# Print first 3 values using next()
total = 0

for i in range(3):
    value = next(checkins)
    print("Day", i + 1, ":", value)
    total += value

# Loop through the remaining values
for day, value in enumerate(checkins, start=4):
    print("Day", day, ":", value)
    total += value

# Print total check-ins
print("\nTotal check-ins:", total)
```

<p>Finish by writing a generator `renewal_alerts(members)` that yields an alert string only for members whose days_left is 7 or less, from a list of 8 member dictionaries you type. Print the generator object, loop through it to print alerts, and print how many were yielded. Run Restart & Run All.</p>

```python
# List of 8 gym members
members = [
    {"name": "Rahul", "days_left": 3},
    {"name": "Priya", "days_left": 15},
    {"name": "Amit", "days_left": 7},
    {"name": "Neha", "days_left": 20},
    {"name": "Karan", "days_left": 5},
    {"name": "Riya", "days_left": 10},
    {"name": "Jay", "days_left": 2},
    {"name": "Pooja", "days_left": 30}
]

# Generator function
def renewal_alerts(members):
    for member in members:
        if member["days_left"] <= 7:
            yield f"Renewal alert: {member['name']} - {member['days_left']} days left!"

# Create generator object
alerts = renewal_alerts(members)

# Print generator object
print("Generator object:", alerts)

# Print renewal alerts
count = 0

for alert in alerts:
    print(alert)
    count += 1

# Print total alerts
print("\nTotal alerts:", count)
```
