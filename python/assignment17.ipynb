<p>In 'recharge_analysis.ipynb' use the `statistics` module on 8 monthly data-usage values in GB you type: print `mean()`, `median()` and `stdev()` rounded to 2 decimals. Also use `math.ceil()` to print how many 28-day recharges cover 100 days.</p>

```python
# Import modules
import statistics
import math

# Monthly data usage values in GB (8 months)
data_usage = [12, 15, 18, 10, 20, 16, 14, 22]

# Calculate mean, median and standard deviation
print("Mean:", round(statistics.mean(data_usage), 2), "GB")
print("Median:", round(statistics.median(data_usage), 2), "GB")
print("Standard Deviation:", round(statistics.stdev(data_usage), 2), "GB")

# Calculate number of 28-day recharges needed for 100 days
recharges = math.ceil(100 / 28)

print("Recharges needed for 100 days:", recharges)
```

<p>Use `datetime.strptime` with format '%d/%m/%Y' to read the recharge date '05/11/2025', add each of 5 plan validities with `timedelta(days=...)`, and print the expiry dates in DD/MM/YYYY using `strftime`. Create the folder 'recharge_out' with `os.makedirs('recharge_out', exist_ok=True)` and save the results as 'recharge_out/expiry_dates.txt'.</p>

```python
# Import modules
from datetime import datetime, timedelta
import os

# Recharge start date
recharge_date = datetime.strptime("05/11/2025", "%d/%m/%Y")

# Five recharge plan validities in days
validities = [28, 30, 56, 84, 90]

# Create output folder
os.makedirs("recharge_out", exist_ok=True)

# Calculate expiry dates and save them in a file
with open("recharge_out/expiry_dates.txt", "w", encoding="utf-8") as file:

    for days in validities:
        expiry_date = recharge_date + timedelta(days=days)

        result = (
            f"Validity: {days} days - "
            f"Expiry Date: {expiry_date.strftime('%d/%m/%Y')}"
        )

        print(result)
        file.write(result + "\n")

print("\nExpiry dates saved successfully!")
```
