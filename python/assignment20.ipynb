```python
class BikeListing:
    def __init__(self, model: str, reserve_price: float):
        self.model = model
        self.reserve_price = reserve_price  # invokes @reserve_price.setter

    @property
    def reserve_price(self) -> float:
        return self.__reserve_price

    @reserve_price.setter
    def reserve_price(self, value: float) -> None:
        if value < 5000 or value > 200000:
            raise ValueError("Reserve price must be between 5000 and 200000.")
        self.__reserve_price = value


listing = BikeListing("Commuter 150", 30000)

test_prices = [30000, 125000, 2500, 250000]  # 2 valid, 2 out-of-range

for price in test_prices:
    try:
        listing.reserve_price = price
        print(f"Price {price}: ACCEPTED (current={listing.reserve_price})")
    except ValueError as err:
        print(f"Price {price}: ERROR - {err}")

# Direct private attribute access still fails due to name mangling:
# print(listing.__reserve_price)
```

<p>Create an abstract base class `Listing(ABC)` with an `@abstractmethod` called `final_price(self)`, using `from abc import ABC, abstractmethod`. Show that creating `Listing()` gives a TypeError (commented out), then make `BikeListing` implement `final_price()` by adding a 2 percent transfer fee.</p>

```python
from abc import ABC, abstractmethod


class Listing(ABC):
    @abstractmethod
    def final_price(self) -> float:
        """Return the final price including applicable fees."""
        pass


# Instantiating an abstract class directly fails with TypeError:
# base_listing = Listing()


class BikeListing(Listing):
    def __init__(self, model: str, reserve_price: float):
        self.model = model
        self.reserve_price = reserve_price

    @property
    def reserve_price(self) -> float:
        return self.__reserve_price

    @reserve_price.setter
    def reserve_price(self, value: float) -> None:
        if value < 5000 or value > 200000:
            raise ValueError("Reserve price must be between 5000 and 200000.")
        self.__reserve_price = value

    def final_price(self) -> float:
        return round(self.reserve_price * 1.02, 2)


listing = BikeListing("Commuter 150", 30000)
print(f"Model: {listing.model}")
print(f"Reserve price: {listing.reserve_price}")
print(f"Final price (with 2% fee): {listing.final_price()}")
```

```python
from abc import ABC, abstractmethod


class Listing(ABC):
    @abstractmethod
    def final_price(self) -> float:
        """Return the final price including applicable fees."""
        pass


# Instantiating an abstract class directly fails with TypeError:
# base_listing = Listing()


class BikeListing(Listing):
    def __init__(self, model: str, reserve_price: float):
        self.model = model
        self.reserve_price = reserve_price

    @property
    def reserve_price(self) -> float:
        return self.__reserve_price

    @reserve_price.setter
    def reserve_price(self, value: float) -> None:
        if value < 5000 or value > 200000:
            raise ValueError("Reserve price must be between 5000 and 200000.")
        self.__reserve_price = value

    def final_price(self) -> float:
        return round(self.reserve_price * 1.02, 2)


listing = BikeListing("Commuter 150", 30000)
test_prices = [0, -100, 500, 5000, 30000, 250000]

for price in test_prices:
    try:
        listing.reserve_price = price
        print(
            f"Price {price}: ACCEPTED | "
            f"Reserve: {listing.reserve_price} | Final Price: {listing.final_price()}"
        )
    except ValueError as err:
        print(f"Price {price}: REJECTED ({err})")
```
