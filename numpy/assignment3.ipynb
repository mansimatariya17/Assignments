Create a NumPy array representing the number of likes on 7 Instagram posts and print its ndim, shape, size, dtype, itemsize, and nbytes properties.

```python
import numpy as np
```

```python
instagram_likes = np.array([450,780,620,970,832,1000,745])

print('instagram likes:',instagram_likes)
print('number of dimensions:', instagram_likes.ndim)
print('Shape:', instagram_likes.shape)
print('size:', instagram_likes.size)
print('Data Type:', instagram_likes.dtype)
print('Item-size:', instagram_likes.itemsize)
print('nbytes:', instagram_likes.nbytes)
```
