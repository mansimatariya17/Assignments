```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Use plt.plot() to create a line chart showing the number of daily steps you walked over the last 7 days. Set the line color to green and linewidth to 2.

```python
steps = [550,400,500,659,700,487,900]
days = ['Day 1','Day 2','Day 3','Day 4','Day 5','Day 6','Day 7']

plt.figure(figsize=(8,5))
plt.plot(days, steps, color='green', linewidth=2, marker='o')
plt.title('Daily Steps Over the Past 7 Days')
plt.xlabel('Days')
plt.ylabel('Steps')
plt.tight_layout()
plt.show()
```

Plot a line chart of your favorite music app's monthly active users for 6 months using plt.plot(). Add circle markers at each data point and set the marker color to red.<br><br><em><strong>Hint:</strong> Use the marker and markerfacecolor parameters in plt.plot().</em>

```python
data = np.random.randint(36985,74589,(6))
users = data.tolist()
month = ['Jan','Feb','Mar','Apr','May', 'Jun']

plt.figure(figsize=(8,5))
plt.plot(month, users, color='orange', marker='o', markerfacecolor='red')
plt.title('Monthly active users for 6 months')
plt.xlabel('Months')
plt.ylabel('Users')
plt.tight_layout()
plt.show()
```

Create a line chart comparing the number of orders you placed on Zomato and Swiggy over the last 5 months. Use different line styles and colors for each app, and add a legend to distinguish them.

```python
Zomato_order = [10,15,20,14,17]
Swiggy_order = [20,25,18,30,10]
month = ['Jan','Feb','Mar','Apr','May']

plt.figure(figsize=(10,4))

plt.plot(month, Zomato_order, color='red', marker='o', label='zomato')
plt.plot(month, Swiggy_order, color='green', linestyle='--', label='swiggy')

plt.title('Zomato and Swiggy order over last 5 month')
plt.xlabel('Months')
plt.ylabel('Orders')
plt.legend(['Zomato','Swiggy'])
plt.tight_layout()
plt.show()
```

Add a main title 'Monthly Data Usage Trend' and label the X-axis as 'Month' and Y-axis as 'Data Used (GB)' to a line chart showing your mobile data usage for the past 6 months.<br><br><em><strong>Hint:</strong> Use plt.title(), plt.xlabel(), and plt.ylabel().</em>

```python
months = ['Jan','Feb','Mar','Apr','May']
data_use = [50,65,45,60,40]

plt.figure(figsize=(9,4))
plt.plot(months, data_use, color='brown', linestyle='-.', marker='o')
plt.title('Monthly data usage Trend')
plt.xlabel('Month')
plt.ylabel('Data Used(GB)')
plt.tight_layout()
plt.show()
```
