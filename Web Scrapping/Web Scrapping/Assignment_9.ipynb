Clean a scraped Flipkart products dataset using pandas by replacing all missing prices with the value 0 and saving the cleaned data as cleaned_products.xlsx.

```python
import pandas as pd

# 1. Load your scraped Flipkart dataset (adjust the filename and format as needed)
input_file = 'flipkart_products.csv'  # Use pd.read_excel() or pd.read_json() if applicable

try:
    df = pd.read_csv(input_file)

    # 2. Clean the price column (replace 'price' with the exact column name in your dataset, e.g., 'Price' or 'Selling Price')
    price_col = 'price'

    # Optional: If your scraped prices include currency symbols or commas (e.g., "₹1,499"),
    # strip them out and convert to numeric before filling missing values:
    df[price_col] = (
        df[price_col]
        .astype(str)
        .str.replace(r'[₹,]', '', regex=True)
        .pipe(pd.to_numeric, errors='coerce')
    )

    # Replace all missing (NaN) prices with 0
    df[price_col] = df[price_col].fillna(0)

    # 3. Save the cleaned DataFrame to an Excel file
    output_file = 'cleaned_products.xlsx'
    df.to_excel(output_file, index=False)

    print(f"Dataset cleaned successfully! Saved {len(df)} rows to '{output_file}'.")

except FileNotFoundError:
    print(f"Error: Could not find '{input_file}'. Please verify the file path.")
except KeyError:
    print(f"Error: Column '{price_col}' not found. Available columns: {list(df.columns)}")
except Exception as e:
    print(f"An error occurred: {e}")
```

Take a scraped Zomato food menu CSV file and remove all rows where the 'rating' column is missing or blank using pandas, then save the cleaned data as menu_cleaned.csv.<br><br><em><strong>Hint:</strong> Use dropna() with the subset parameter for the 'rating' column.</em>

```python
import pandas as pd
import numpy as np

# 1. Load the scraped Zomato food menu dataset
input_file = 'zomato_menu.csv'  # Replace with your actual filename if different

try:
    df = pd.read_csv(input_file)
    print(f"Original dataset loaded with {len(df)} rows.")

    # 2. Standardize blank/whitespace-only entries in the 'rating' column into true NaNs
    df['rating'] = df['rating'].replace(r'^\s*$', np.nan, regex=True)

    # 3. Use dropna() with the subset parameter to remove rows where 'rating' is missing
    cleaned_df = df.dropna(subset=['rating'])

    # 4. Export the cleaned data to a new CSV file
    output_file = 'menu_cleaned.csv'
    cleaned_df.to_csv(output_file, index=False)

    print(f"Success! Removed missing ratings. Cleaned data saved to '{output_file}' with {len(cleaned_df)} rows remaining.")

except FileNotFoundError:
    print(f"Error: The file '{input_file}' was not found.")
except KeyError:
    print("Error: The 'rating' column was not found in the dataset. Please check your column headers.")
except Exception as e:
    print(f"An error occurred: {e}")
```

Use pandas to combine two scraped datasets: one of Spotify playlists (CSV) and one of song details (Excel), then save the merged data as merged_playlist_songs.csv.<br><br><em><strong>Hint:</strong> Use merge() and ensure you join on the song name or ID.</em>

```python
import pandas as pd

# File paths for your datasets
playlists_file = 'spotify_playlists.csv'
songs_file = 'song_details.xlsx'

try:
    # 1. Load the Spotify playlists dataset (CSV)
    print(f"Loading playlists from '{playlists_file}'...")
    playlists_df = pd.read_csv(playlists_file)

    # 2. Load the song details dataset (Excel)
    print(f"Loading song details from '{songs_file}'...")
    songs_df = pd.read_excel(songs_file)

    # 3. Merge the two datasets
    # Replace 'song_id' with the exact common column name in both datasets (e.g., 'song_id' or 'track_name')
    merge_key = 'song_id' 
    
    # Using an 'inner' merge by default; you can change to 'left', 'right', or 'outer' based on your needs
    print(f"Merging datasets on column: '{merge_key}'...")
    merged_df = pd.merge(playlists_df, songs_df, on=merge_key, how='inner')

    # 4. Save the merged DataFrame to a new CSV file
    output_filename = 'merged_playlist_songs.csv'
    merged_df.to_csv(output_filename, index=False)

    print(f"\nSuccess! Merged dataset contains {len(merged_df)} rows and has been saved to '{output_filename}'.")

except FileNotFoundError as e:
    print(f"Error: Could not find the file. Please check your file paths. Details: {e}")
except KeyError as e:
    print(f"Error: The merge key or column was not found. Please check your column names. Details: {e}")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```
