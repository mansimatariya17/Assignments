```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Create a simple bar chart showing the number of orders for 5 different Zomato restaurants and save the chart as a PNG file using plt.savefig() with a DPI of 150.

```python
restaurants = ['Dominos', 'McDonalds', 'Pizza Hut', 'Burger King', 'Subway']
orders = [450, 380, 320, 290, 250]

plt.figure(figsize=(8, 5))
plt.bar(restaurants, orders, color="#3B286E")
plt.title('Number of Orders by Restaurant')
plt.tight_layout()

# Save chart as a PNG file at 150 DPI
plt.savefig('zomato_restaurant_orders.png', dpi=150)

plt.show()
```

Generate a line plot that displays your daily step count for one week (7 days) and export it as an SVG file using plt.savefig().<br><br><em><strong>Hint:</strong> Change the file extension in the filename argument to save as SVG.</em>

```python
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
steps = [4500, 6200, 5800, 7500, 8200, 10000, 9000]

# Create line plot
plt.figure(figsize=(8, 5))
plt.plot(days, steps, marker='o', color='blue', linewidth=2)

# Add title and labels
plt.title('Daily Step Count for One Week')
plt.xlabel('Day')
plt.ylabel('Number of Steps')

plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()

# Export as an SVG file
plt.savefig('weekly_step_count.svg', format='svg')

# Display the plot
plt.show()
```

Build a pie chart showing the percentage of time you spend on 4 different apps (Instagram, YouTube, WhatsApp, Spotify) in a day and save the chart as a PDF using plt.savefig().

```python
apps = ['Instagram', 'YouTube', 'WhatsApp', 'Spotify']
hours = [2, 3, 1.5, 1.5]

# Create pie chart
plt.figure(figsize=(7, 7))

plt.pie(
    hours,
    labels=apps,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Daily Time Spent on Apps')

# Save chart as a PDF file
plt.savefig('daily_app_usage.pdf', format='pdf', bbox_inches='tight')

plt.show()
```
