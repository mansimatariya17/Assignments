Install BeautifulSoup and requests in your Python environment, then write a script to fetch and print the HTML content of the Flipkart homepage.

```python
import requests
from bs4 import BeautifulSoup

# Define the Flipkart URL
url = "https://www.flipkart.com"

# Add headers to mimic a real browser request
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                  "AppleWebKit/537.36 (KHTML, like Gecko) "
                  "Chrome/120.0.0.0 Safari/537.36"
}

try:
    # Send GET request to Flipkart
    response = requests.get(url, headers=headers)
    response.raise_for_status()
    
    # Parse the HTML content using BeautifulSoup
    soup = BeautifulSoup(response.text, 'html.parser')
    
    # Print the pretty-printed HTML content (or slice it to see a portion)
    print("--- Flipkart HTML Content (First 1000 characters) ---")
    print(soup.prettify()[:1000])

except requests.exceptions.RequestException as e:
    print(f"An error occurred: {e}")
```

Use BeautifulSoup's find() method to extract and print the first product title from a sample HTML file that contains a list of Zomato restaurant cards.

```python
from bs4 import BeautifulSoup

# Sample HTML containing Zomato-style restaurant cards
sample_html = """
<div class="search-results-container">
    <div class="restaurant-card">
        <h3 class="restaurant-name">The Belgian Waffle Co.</h3>
        <p class="cuisine">Desserts, Beverages</p>
    </div>
    <div class="restaurant-card">
        <h3 class="restaurant-name">Mezbaan Family Restaurant</h3>
        <p class="cuisine">Biryani, North Indian</p>
    </div>
</div>
"""

# Parse the HTML content using BeautifulSoup
soup = BeautifulSoup(sample_html, 'html.parser')

# Use soup.find() to locate the very first matching element
first_title_element = soup.find('h3', class_='restaurant-name')

# Extract and print the text if the element is found
if first_title_element:
    print("First Restaurant Title:", first_title_element.get_text().strip())
else:
    print("No matching element found.")
```

Write a script that uses find_all() to collect all image tags from the homepage of Myntra and prints the src attribute of each image.<br><br><em><strong>Hint:</strong> Use requests to get the HTML, then BeautifulSoup to parse and extract the src values.</em>

```python
import requests
from bs4 import BeautifulSoup

# Target URL for Myntra
url = "https://www.myntra.com"

# Headers to mimic a real browser request and prevent 403 Forbidden blocking
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                  "AppleWebKit/537.36 (KHTML, like Gecko) "
                  "Chrome/120.0.0.0 Safari/537.36"
}

try:
    # Send GET request to Myntra
    response = requests.get(url, headers=headers)
    response.raise_for_status()
    
    # Parse the HTML content using BeautifulSoup
    soup = BeautifulSoup(response.text, 'html.parser')
    
    # Use find_all() to collect all <img> tags
    image_tags = soup.find_all('img')
    
    print(f"Total image tags found: {len(image_tags)}\n")
    print("--- Image Sources (src) ---")
    
    # Iterate and extract the 'src' attribute for each image tag
    for index, img in enumerate(image_tags, start=1):
        # .get('src') safely returns the attribute value or None if not present
        src_value = img.get('src')
        
        # Modern sites often use lazy loading where the real URL is stored in 'data-src'
        # We can also check for 'data-src' as a fallback
        if not src_value:
            src_value = img.get('data-src', 'No src found')
            
        print(f"{index}. {src_value}")

except requests.exceptions.RequestException as e:
    print(f"An error occurred: {e}")
```

Use BeautifulSoup's select() method with a CSS selector to extract all anchor tags inside a div with class 'trending' from a given HTML snippet (simulate a 'Trending Now' section like YouTube). Print the href attribute of each link.

```python
from bs4 import BeautifulSoup

# Sample HTML snippet simulating a YouTube-style 'Trending Now' section
sample_html = """
<div class="main-content">
    <h2>Trending Now</h2>
    <div class="trending">
        <a href="/watch?v=dQw4w9WgXcQ">Never Gonna Give You Up</a>
        <a href="/watch?v=9bZkp7q19f0">Gangnam Style</a>
        <a href="/watch?v=kJQP7kiw5Fk">Despacito</a>
    </div>
    <div class="other-section">
        <a href="/watch?v=unrelated123">Unrelated Video</a>
    </div>
</div>
"""

# Parse the HTML content using BeautifulSoup
soup = BeautifulSoup(sample_html, 'html.parser')

# Use select() with the CSS selector 'div.trending a' to target <a> tags inside <div class="trending">
trending_links = soup.select('div.trending a')

print(f"Total trending links found: {len(trending_links)}\n")
print("--- Trending Link Hrefs ---")

# Iterate through the selected elements and print their href attributes
for index, link in enumerate(trending_links, start=1):
    href_value = link.get('href')
    link_text = link.get_text().strip()
    print(f"{index}. Text: '{link_text}' -> Href: {href_value}")
```

Extract and print all the text content of the first 5 movie titles from https://quotes.toscrape.com/ using BeautifulSoup.<br><br><em><strong>Constraint:</strong> Only print the text, not the HTML tags or attributes.</em>

```python
import requests
from bs4 import BeautifulSoup

# Target URL
url = "https://quotes.toscrape.com/"

try:
    # Send a GET request to the website
    response = requests.get(url)
    response.raise_for_status()
    
    # Parse the HTML content using BeautifulSoup
    soup = BeautifulSoup(response.text, 'html.parser')
    
    # Find all elements containing the quote text (<span class="text">)
    quote_elements = soup.find_all('span', class_='text')
    
    print("First 5 Quotes extracted:\n")
    
    # Slice the list to get only the first 5 elements and print their pure text content
    for index, quote in enumerate(quote_elements[:5], start=1):
        # .get_text() extracts only the text, stripping away all HTML tags and attributes
        print(f"{index}. {quote.get_text().strip()}")

except requests.exceptions.RequestException as e:
    print(f"An error occurred: {e}")
```
