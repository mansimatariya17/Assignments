
<p>Open 'coaching_marks.ipynb' and create the list students with 10 dictionaries holding name and marks out of 150. Use `map()` with a lambda to build percentages rounded to 1 decimal and print `list()` of the result. Hint: `map()` returns an iterator, so wrap it in `list()` to see the values. If you missed the earlier session, start a new notebook with this list.</p>

```python
students = [
    {"name": "Amit", "marks": 135},
    {"name": "Priya", "marks": 140},
    {"name": "Rahul", "marks": 120},
    {"name": "Neha", "marks": 145},
    {"name": "Karan", "marks": 110},
    {"name": "Riya", "marks": 130},
    {"name": "Jay", "marks": 125},
    {"name": "Pooja", "marks": 115},
    {"name": "Ankit", "marks": 138},
    {"name": "Meera", "marks": 148}
]

percentages = map(
    lambda student: round((student["marks"] / 150) * 100, 1),
    students
)

print(list(percentages))
```

<p>Use `filter()` with a lambda to keep students above 75 percent and print the count and names. Then use `sorted()` with `key=lambda s: s['marks']` and `reverse=True` to print the top 3 students by marks.</p>

```python
above_75 = list(filter(
    lambda s: (s["marks"] / 150) * 100 > 75,
    students
))

print("Count of students:", len(above_75))
print("Student names:")

for s in above_75:
    print(s["name"])
```

```python
# sort students by marks in descending order:
sorted_students = sorted(
    students,
    key=lambda s: s["marks"],
    reverse=True
)

print("Top 3 Students:")

for s in sorted_students[:3]:
    print(s["name"], "-", s["marks"], "marks")
```

<p>Import `reduce` from `functools` and use it with lambdas on a list of 10 fee payments you type: first to get the total fees collected, then to find the highest single payment without using `max()`. Print both results.</p>

```python
from functools import reduce

fees = [5000, 6500, 7000, 5500, 8000,
        6000, 7500, 9000, 4500, 8500]

total_fees = reduce(
    lambda a, b: a + b,
    fees
)

print("Total fees collected:", total_fees)

highest_payment = reduce(
    lambda a, b: a if a > b else b,
    fees
)

print("Highest single payment:", highest_payment)
```

<p>Finish by defining `apply_to_all(func, items)` that returns a list, and use it with three different lambdas: names in title case, marks plus 5 grace marks capped at 150 using `min()`, and percentages converted to grades by calling `assign_grade`. Print all three outputs and run Restart & Run All.</p>

```python
def assign_grade(p):
    if p >= 90:
        return "A"
    elif p >= 75:
        return "B"
    elif p >= 60:
        return "C"
    else:
        return "D"


def apply_to_all(func, items):
    return list(map(func, items))


# 1. Names in title case
names = apply_to_all(lambda s: s["name"].title(), students)

# 2. Add 5 grace marks, maximum 150
grace_marks = apply_to_all(lambda s: min(s["marks"] + 5, 150), students)

# 3. Convert percentages into grades
grades = apply_to_all(
    lambda s: assign_grade((s["marks"] / 150) * 100),
    students
)

print("Names:", names)
print("Grace Marks:", grace_marks)
print("Grades:", grades)
```
