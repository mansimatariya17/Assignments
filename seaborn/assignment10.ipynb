```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Use sns.set() to apply the 'poster' context to a Seaborn barplot showing the number of songs streamed per day on Spotify, and observe how the plot elements change compared to the default context.

```python
df = pd.DataFrame({
    'day': ['Monday', 'Tuesday', 'Wednesday', 'Thursday',
            'Friday', 'Saturday', 'Sunday'],
    'songs_streamed': [120, 150, 100, 180, 220, 300, 250]
})

# Apply poster context
plt.figure(figsize=(18,10))
sns.set_context('poster')

# Create barplot
sns.barplot(data=df, x='day', y='songs_streamed')

# Add title and labels
plt.title('Spotify Songs Streamed Per Day')
plt.xlabel('Day')
plt.ylabel('Number of Songs Streamed')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

Create a Seaborn countplot showing the distribution of food orders by cuisine type (e.g., Indian, Chinese, Italian) using a 'colorblind' palette, similar to how Zomato might visualize order trends.<br><br><em><strong>Hint:</strong> Use palette='colorblind' in your plot function.</em>

```python
df = pd.DataFrame({
    'cuisine_type': [
        'Indian', 'Chinese', 'Italian', 'Indian', 'Mexican',
        'Chinese', 'Indian', 'Italian', 'Indian', 'Chinese',
        'Italian', 'Indian', 'Mexican', 'Chinese', 'Indian',
        'Italian', 'Chinese', 'Indian', 'Mexican', 'Italian'
    ]
})

# Set the Seaborn theme
sns.set_theme(style='whitegrid')

# Create countplot using colorblind palette
sns.countplot(data=df, x='cuisine_type', palette='colorblind', hue='cuisine_type', legend=False)

# Add title and labels
plt.title('Food Orders by Cuisine Type - Zomato')
plt.xlabel('Cuisine Type')
plt.ylabel('Number of Orders')

plt.tight_layout()
plt.show()
```

Generate a Seaborn lineplot of daily step counts (simulate data for 2 weeks) and style it using the 'muted' palette and 'notebook' context for a clean, presentation-ready look.

```python
df = pd.DataFrame({
    'day': range(1, 15),
    'steps': [
        4500, 6200, 5800, 7500, 8100, 6900, 9200,
        8500, 7200, 9800, 10500, 8900, 11200, 10000
    ]
})

# Apply muted color palette and notebook context
sns.set_theme(style='whitegrid', palette='muted', context='notebook')

# Create lineplot
sns.lineplot(data=df, x='day', y='steps', marker='o')

# Customize the plot
plt.title('Daily Step Counts Over Two Weeks')
plt.xlabel('Day')
plt.ylabel('Number of Steps')
plt.xticks(range(1, 15))

plt.tight_layout()
plt.show()
```

Refactor an existing Seaborn plot by switching its palette from 'deep' to 'colorblind' and changing the context from 'talk' to 'poster'. Briefly describe the visual differences you notice after applying these changes.

```python
df = pd.DataFrame({
    'cuisine': ['Indian', 'Chinese', 'Italian', 'Mexican', 'Indian',
                'Chinese', 'Italian', 'Mexican', 'Indian', 'Chinese'],
    'orders': [120, 90, 75, 60, 150, 110, 85, 70, 130, 100]
})

# Apply updated palette and context
plt.figure(figsize=(18,10))
sns.set_theme(style='whitegrid', palette='colorblind', context='poster')

# Create barplot
sns.barplot(data=df, x='cuisine', y='orders', errorbar=None)

# Customize the plot
plt.title('Food Orders by Cuisine')
plt.xlabel('Cuisine Type')
plt.ylabel('Number of Orders')

plt.tight_layout()
plt.show()
```

Design a Seaborn theme that mimics the look of a Flipkart sales dashboard: use sns.set() to set a 'deep' palette, 'talk' context, and customize the font scale and grid style. Apply your theme to a barplot showing sales of 5 product categories.

```python
df = pd.DataFrame({
    'product_category': [
        'Mobiles', 'Fashion', 'Electronics',
        'Home Appliances', 'Books'
    ],
    'sales': [150000, 95000, 180000, 120000, 45000]
})

# Customize Seaborn theme
sns.set(
    context='talk',
    style='whitegrid',
    palette='deep',
    font_scale=1.1,
    rc={'grid.linestyle': '--', 'grid.alpha': 0.5}
)

# Create barplot
plt.figure(figsize=(10, 6))

sns.barplot(
    data=df,
    x='product_category',
    y='sales',
    hue='product_category',
    legend=False
)

# Customize the plot
plt.title('Flipkart Sales by Product Category', pad=15)
plt.xlabel('Product Category')
plt.ylabel('Sales (₹)')
plt.xticks(rotation=20)

plt.tight_layout()
plt.show()
```
