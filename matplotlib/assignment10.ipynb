```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Create a line chart using Matplotlib to show the number of daily active users for a music streaming app (like Spotify) over 12 months, and use plt.annotate() to label the month with the highest user count.

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
          'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

active_users = [85, 90, 88, 95, 100, 105,
                110, 108, 115, 120, 118, 130]

# Find the month with the highest users
max_users = max(active_users)
max_index = active_users.index(max_users)
max_month = months[max_index]

# Create line chart
plt.figure(figsize=(10, 5))
plt.plot(months, active_users, marker='o',
         color='green', linewidth=2)

# Annotate the highest user count
plt.annotate(
    f'Peak: {max_users} million ({max_month})',
    xy=(max_month, max_users),
    xytext=(-80, -30),
    textcoords='offset points',
    arrowprops=dict(arrowstyle='->', color='red'),
    color='red',
    fontsize=10
)

plt.grid()
plt.show()
```

Plot a bar chart showing the number of orders per month for a food delivery app (like Zomato), and add an arrow annotation using plt.annotate() to highlight the month with the lowest orders.<br><br><em><strong>Hint:</strong> Use the 'arrowprops' parameter to draw the arrow pointing to the bar.</em>

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
          'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

orders = [1200, 950, 1100, 1400, 1300, 800,
          1500, 1600, 1250, 1700, 1450, 1800]

# Find the month with the lowest orders
min_orders = min(orders)
min_index = orders.index(min_orders)
min_month = months[min_index]

# Create bar chart
plt.figure(figsize=(10, 5))
plt.bar(months, orders, color="#D63E3E", edgecolor='black')

# Add arrow annotation to the lowest bar
plt.annotate(
    f'Lowest Orders: {min_orders}',
    xy=(min_index, min_orders),
    xytext=(min_index + 1, min_orders + 400),
    arrowprops=dict(arrowstyle='->', color='blue', lw=2),
    color="blue",
    fontsize=10,
    ha='center'
)

plt.show()
```

Create a pie chart representing the percentage of different payment methods (UPI, Card, Wallet, Cash) used on an e-commerce app (like Flipkart), and add a legend positioned at the lower right corner of the plot.

```python
payment_methods = ['UPI', 'Card', 'Wallet', 'Cash']
percentages = [45, 30, 15, 10]

# Create pie chart
plt.figure(figsize=(7, 7))

plt.pie(
    percentages,
    labels=payment_methods,
    autopct='%1.1f%%',
    startangle=90
)

# Add legend at the lower-right corner
plt.legend(
    payment_methods,
    loc='upper right',
    title='Payment Methods'
)

plt.title('E-commerce App Payment Methods')
plt.show()
```

Given a scatter plot of movie ratings vs. number of reviews for movies on a streaming platform (like BookMyShow), use plt.annotate() to add a text box highlighting the movie with the highest rating.<br><br><em><strong>Hint:</strong> Set the 'bbox' parameter in plt.annotate() to create a visible text box.</em>

```python
movies = ['Movie A', 'Movie B', 'Movie C', 'Movie D', 'Movie E']
ratings = [8.2, 9.1, 7.8, 9.5, 8.7]
reviews = [1200, 2500, 900, 3200, 1800]

# Find the movie with the highest rating
max_rating = max(ratings)
max_index = ratings.index(max_rating)

# Create scatter plot
plt.figure(figsize=(8, 5))
plt.scatter(reviews, ratings, color='purple', s=100)

# Annotate the highest-rated movie with a text box
plt.annotate(
    f'{movies[max_index]}\nRating: {max_rating}',
    xy=(reviews[max_index], ratings[max_index]),
    xytext=(reviews[max_index] - 1000, ratings[max_index] - 0.5),
    bbox=dict(boxstyle='round,pad=0.5',
              fc='yellow', alpha=0.7),
    arrowprops=dict(arrowstyle='->', color='red')
)

plt.grid()
plt.tight_layout()
plt.show()
```
