Given a pandas DataFrame of Flipkart product reviews with columns: product_id, user_id, rating, review_text, identify and print the number of duplicate reviews (where product_id and user_id are both the same) before and after using drop_duplicates().

```python
df = pd.DataFrame({
    'product_id': [101, 102, 101, 103, 102, 104, 101],
    'user_id': [1, 2, 1, 3, 2, 4, 5],
    'rating': [5, 4, 5, 3, 4, 5, 2],
    'review_text': ['Excellent product','Good quality','Excellent product','Average product','Good quality','Very useful','Not satisfied'
    ]
})

duplicated_before = df.duplicated(subset=['product_id','user_id']).sum()
print('Duplicates Before:',duplicated_before)

df = df.drop_duplicates(subset=['product_id','user_id'],keep='first')

duplicated_after = df.duplicated(subset=['product_id','user_id']).sum()
print('Duplicates After:',duplicated_after)
```

Use the IQR method to detect outliers in the 'order_amount' column of a Zomato orders DataFrame and print the order IDs that are considered outliers.<br><br><em><strong>Hint:</strong> Calculate Q1, Q3, and IQR, then filter orders outside the [Q1 - 1.5*IQR, Q3 + 1.5*IQR] range.</em>

```python
data = {
    'transaction_id': [1, 2, 3, 4, 5, 6, 7, 8],
    'transaction_amount': [500, 700, 450, 600, 550, 650, 500, 10000]
}

q1 = np.quantile(data['transaction_amount'],0.25)

q3 = np.quantile(data['transaction_amount'],0.75)

iqr = q3 - q1

lower_bound = q1 - (1.5 * iqr)
upper_bound = q3 + (1.5 * iqr)

print('lower bound',lower_bound)
print('upper bound',upper_bound)
```

For a DataFrame containing daily IPL match ticket sales (columns: date, tickets_sold), calculate the Z-score for each day's tickets_sold, and print the dates where the absolute Z-score is greater than 2.<br><br><em><strong>Constraint:</strong> Use pandas and scipy.stats.zscore for calculations.</em>

```python
data = {
    'date': pd.date_range(start='2026-04-01', periods=10, freq='D'),
    'tickets_sold': [15860, 18772, 18092, 15466, 19426, 35000, 18171, 17919, 3000, 16685]
}

df = pd.DataFrame(data)

# 2. Calculate Z-score using scipy.stats.zscore
df['z_score'] = stats.zscore(df['tickets_sold'])

# 3. Filter and print dates where absolute Z-score > 2
outliers = df[df['z_score'].abs() > 2]

print("--- Outlier Dates (|Z-score| > 2) ---")
print(outliers[['date', 'tickets_sold', 'z_score']])
```

Use ChatGPT or Copilot to generate a pandas function that finds and returns all outlier transaction amounts in a Paytm wallet DataFrame using the Z-score method, then test the function with sample data.

```python
def find_outliers(df, column='transaction_amount', threshold=2):
    # Calculate Z-scores
    mean = df[column].mean()
    std = df[column].std()

    z_scores = (df[column] - mean) / std

    # Return transactions with |Z-score| > 3
    return df[abs(z_scores) > threshold]


# Sample Paytm wallet transaction data
data = {
    'transaction_id': [1, 2, 3, 4, 5, 6, 7, 8],
    'transaction_amount': [500, 700, 450, 600, 550, 650, 500, 10000]
}

df = pd.DataFrame(data)

# Find outliers
outliers = find_outliers(df)

print("Outlier Transactions:")
print(outliers)
```
