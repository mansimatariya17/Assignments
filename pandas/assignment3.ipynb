Given a DataFrame of Flipkart products with columns 'product_name', 'category', and 'price', use iloc to select the first 10 rows and all columns, and print the result.

```python
df = pd.DataFrame({
    'product_name': ['Phone', 'Laptop', 'Headphones', 'Watch', 'TV',
                     'Tablet', 'Camera', 'Keyboard', 'Mouse', 'Speaker'],
    'category': ['Electronics'] * 10,
    'price': [15000, 55000, 2000, 3000, 40000, 20000, 30000, 1500, 800, 2500]
})

result = df.iloc[:10, :]
print(result)
```


Using a DataFrame of food orders from Zomato with columns 'restaurant', 'order_amount', and 'delivery_time', filter and display only those orders where 'order_amount' is greater than 500 using a boolean condition.

```python
df = pd.DataFrame({
    'restaurant': ['Dominos', 'McDonalds', 'Subway', 'Pizza Hut', 'Burger King','KFC'],
    'order_amount': [450, 650, 300, 800, 550, 600],
    'delivery_time': [30, 25, 40, 35, 20, 15]
})

result = df[df['order_amount'] > 500]
print(result)
```

Given a DataFrame of Spotify songs with columns 'song', 'artist', 'streams', and 'duration', use the query() method to select all songs with more than 1,000,000 streams and duration less than 180 seconds.<br><br><em><strong>Hint:</strong> Use the query syntax: query('streams > 1000000 and duration < 180')</em>

```python
df = pd.DataFrame({
    'song': ['Shape of You', 'Perfect', 'Blinding Lights', 'Stay', 'Bad Guy','Believer'],
    'artist': ['Artist 1', 'Artist 2', 'Artist 3', 'Artist 4', 'Artist 5', 'Artist 6'],
    'streams': [1500000, 800000, 2000000, 1200000, 900000, 1900000],
    'duration': [170, 160, 200, 175, 150, 165]
})

result = df.query('streams > 1000000 and duration < 180')
print(result)
```


Suppose you have a DataFrame of IPL cricket matches with columns 'team1', 'team2', 'winner', and 'total_runs'. Write code to filter and display all matches where 'total_runs' is between 180 and 220 using boolean indexing.<br><br><em><strong>Constraint:</strong> Do not use the query() method for this task.</em>

```python
df = pd.DataFrame({
    'team1': ['MI', 'CSK', 'RCB', 'KKR', 'GT'],
    'team2': ['CSK', 'RCB', 'KKR', 'GT', 'MI'],
    'winner': ['MI', 'CSK', 'KKR', 'KKR', 'GT'],
    'total_runs': [190, 175, 210, 225, 200]
})

result = df[(df['total_runs']>180)&(df['total_runs']<220)]
print(result)
```
