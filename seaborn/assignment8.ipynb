```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Use sns.scatterplot() to visualize the relationship between the number of followers and the number of posts for 20 fictional Instagram users. Label the axes appropriately.

```python
data = {
    'followers': [
        500, 1200, 800, 2500, 4000,
        1500, 6000, 3500, 900, 7500,
        2000, 5000, 300, 8500, 4500,
        1100, 6500, 3000, 9500, 1800
    ],
    'posts': [
        50, 120, 80, 200, 350,
        100, 500, 280, 70, 600,
        180, 400, 30, 700, 320,
        90, 550, 250, 800, 140
    ]
}

df = pd.DataFrame(data)

# Create scatter plot
sns.scatterplot(
    data=df,
    x='followers',
    y='posts'
)

# Label the axes and title
plt.xlabel('Number of Followers')
plt.ylabel('Number of Posts')
plt.title('Instagram Followers vs Number of Posts')

plt.show()
```

Create a line plot using sns.lineplot() to show the daily number of food orders on Zomato for each day over a month. Use a list of 30 random order counts to simulate the data.

```python
np.random.seed(42)
orders = np.random.randint(100, 500, 30)

# Create DataFrame
df = pd.DataFrame({
    'Day': range(1, 31),
    'Orders': orders
})

# Create line plot
sns.lineplot(data=df, x='Day', y='Orders', marker='o')

# Add title and labels
plt.title('Daily Zomato Food Orders Over One Month')
plt.xlabel('Day of the Month')
plt.ylabel('Number of Food Orders')

plt.xticks(range(1, 31, 2))
plt.tight_layout()
plt.show()
```

Using sns.relplot(), plot the relationship between movie ratings and the number of reviews for different genres (Action, Comedy, Drama) based on a sample dataset of 15 movies. Use the 'hue' parameter to differentiate genres.<br><br><em><strong>Hint:</strong> Create a DataFrame with columns: 'movie', 'rating', 'reviews', and 'genre'.</em>

```python
data = {
    'movie': [
        'Movie 1', 'Movie 2', 'Movie 3', 'Movie 4', 'Movie 5',
        'Movie 6', 'Movie 7', 'Movie 8', 'Movie 9', 'Movie 10',
        'Movie 11', 'Movie 12', 'Movie 13', 'Movie 14', 'Movie 15'
    ],
    'rating': [
        8.2, 7.5, 6.9, 8.7, 7.8,
        6.5, 7.2, 8.0, 7.6, 6.8,
        9.0, 8.4, 7.1, 7.9, 8.6
    ],
    'reviews': [
        1500, 900, 700, 2000, 1200,
        500, 800, 1800, 1100, 600,
        2500, 1900, 750, 1300, 2200
    ],
    'genre': [
        'Action', 'Action', 'Action', 'Action', 'Action',
        'Comedy', 'Comedy', 'Comedy', 'Comedy', 'Comedy',
        'Drama', 'Drama', 'Drama', 'Drama', 'Drama'
    ]
}

df = pd.DataFrame(data)

# Create relational plot
sns.relplot(
    data=df,
    x='rating',
    y='reviews',
    hue='genre',
    kind='scatter',
    s=100
)

plt.title('Movie Ratings vs Number of Reviews')
plt.xlabel('Movie Rating')
plt.ylabel('Number of Reviews')

plt.show()
```

Generate a regression plot using sns.regplot() to analyze the correlation between the number of hours spent on YouTube and exam scores for a group of 25 students. Interpret whether the trend is positive or negative.

```python
data = {
    'youtube_hours': [
        1, 2, 3, 4, 5, 1.5, 2.5, 3.5, 4.5, 5.5,
        6, 2, 3, 4, 5, 6.5, 7, 1, 2.8, 3.8,
        4.8, 5.8, 6.8, 7.5, 8
    ],
    'exam_scores': [
        95, 90, 85, 80, 75, 93, 88, 82, 78, 72,
        68, 91, 86, 81, 76, 65, 60, 97, 87, 79,
        74, 70, 63, 58, 55
    ]
}

df = pd.DataFrame(data)

# Create regression plot
sns.regplot(
    data=df,
    x='youtube_hours',
    y='exam_scores',
    ci=None,
    scatter_kws={'color': 'blue'},
    line_kws={'color': 'red'}
)

plt.title('YouTube Hours vs Exam Scores')
plt.xlabel('Hours Spent on YouTube per Day')
plt.ylabel('Exam Scores (%)')
plt.show()

# Calculate correlation
print("Correlation:", round(
    df['youtube_hours'].corr(df['exam_scores']), 2
))
```
