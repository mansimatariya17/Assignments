```python
import requests

def fetch_first_10_shows():
    url = "https://api.tvmaze.com/shows"
    
    try:
        print("Fetching TV shows from TVMaze API...")
        response = requests.get(url, timeout=10)
        
        # Raise an exception for HTTP errors (e.g., 404 or 500)
        response.raise_for_status()
        
        # Parse JSON response into a Python list of dictionaries
        shows = response.json()
        
        # Slice the list to get the first 10 shows
        first_10_shows = shows[:10]
        
        print("\n--- First 10 TV Shows ---")
        for index, show in enumerate(first_10_shows, start=1):
            print(f"{index}. {show.get('name')}")
            
    except requests.exceptions.RequestException as e:
        print(f"An error occurred during the request: {e}")

if __name__ == "__main__":
    fetch_first_10_shows()
```

Write a function getWeather(city) that fetches current weather data for a city from the MetaWeather API (https://www.metaweather.com/api/) and prints the temperature. Use 'Ahmedabad' as the test city.<br><br><em><strong>Hint:</strong> You may need to first get the WOEID for the city, then fetch weather using that ID.</em>

```python
import requests

def get_weather(city):
    try:
        # Step 1: Get latitude and longitude for the city using Open-Meteo's geocoding API
        geo_url = f"https://geocoding-api.open-meteo.com/v1/search?name={city}&count=1"
        print(f"Searching for coordinates of {city}...")
        geo_response = requests.get(geo_url, timeout=10)
        geo_response.raise_for_status()
        geo_data = geo_response.json()

        if not geo_data.get("results"):
            print(f"City '{city}' not found.")
            return

        location = geo_data["results"][0]
        lat = location["latitude"]
        lon = location["longitude"]
        city_name = location["name"]
        country = location.get("country", "")

        # Step 2: Fetch current weather using the retrieved coordinates
        weather_url = f"https://api.open-meteo.com/v1/forecast?latitude={lat}&longitude={lon}&current_weather=true"
        print(f"Fetching weather data...")
        weather_response = requests.get(weather_url, timeout=10)
        weather_response.raise_for_status()
        weather_data = weather_response.json()

        # Extract current weather metrics
        current = weather_data.get("current_weather", {})
        temp = current.get("temperature")
        windspeed = current.get("windspeed")

        print(f"\n--- Current Weather for {city_name}, {country} ---")
        print(f"Temperature: {temp}°C")
        print(f"Wind Speed: {windspeed} km/h")

    except requests.exceptions.RequestException as e:
        print(f"An error occurred during the request: {e}")

# Execute the function with 'Ahmedabad'
if __name__ == "__main__":
    get_weather('Ahmedabad')
```

Find a public API that requires an API key (for example, https://newsapi.org or https://openweathermap.org). Register for a free key, and write code to fetch data from one endpoint using your key. Print one interesting field from the response.<br><br><em><strong>Constraint:</strong> Do not share your API key in public code or screenshots.</em>

```python
import os
import requests

# Recommended: Load your API key securely from an environment variable
# (e.g., set WEATHER_API_KEY in your terminal or .env file)
# Alternatively, replace os.getenv(...) with your literal string ONLY for local testing.
API_KEY = os.getenv("WEATHER_API_KEY", "YOUR_API_KEY_HERE")

def get_current_weather(city_name):
    # Construct the API endpoint URL with query parameters
    url = f"https://api.openweathermap.org/data/2.5/weather"
    
    params = {
        "q": city_name,
        "appid": API_KEY,
        "units": "metric"  # Use metric for Celsius
    }
    
    print(f"Fetching weather data for {city_name}...")
    response = requests.get(url, params=params)
    
    # Check if the request was successful
    if response.status_code == 200:
        data = response.json()
        
        # Extract interesting fields from the JSON response
        city = data.get("name")
        country = data.get("sys", {}).get("country")
        temp = data["main"]["temp"]
        weather_description = data["weather"][0]["description"].capitalize()
        
        print(f"\n--- Weather Report for {city}, {country} ---")
        print(f"Temperature: {temp}°C")
        print(f"Condition: {weather_description}")
        
    elif response.status_code == 401:
        print("Error: Unauthorized. Please check if your API key is valid.")
    else:
        print(f"Failed to retrieve data. Status code: {response.status_code}")
        print(response.json())

# Test the function with a city
if __name__ == "__main__":
    get_current_weather("London")
```
