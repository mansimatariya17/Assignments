Create a stacked bar chart using Matplotlib that shows the number of food orders from Zomato and Swiggy for each day of the week. Use different colors for each platform and add a legend.

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

# Number of orders
zomato_orders = [5, 7, 6, 8, 10, 12, 9]
swiggy_orders = [4, 6, 5, 7, 8, 11, 10]

# Create stacked bar chart
plt.bar(days, zomato_orders, color="#5E4A9C", label="Zomato")
plt.bar(days, swiggy_orders, bottom=zomato_orders,color="#6BAACE", label="Swiggy")
plt.legend()
plt.show()
```

Plot an area chart that visualizes the daily active users on Spotify and YouTube over a 7-day period using Matplotlib's fill_between(). Make sure each platform's area is filled with a different color and add labels for clarity.

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

# Daily active users (fictional sample data)
spotify_users = [1200, 1350, 1280, 1500, 1650, 1900, 1750]
youtube_users = [2500, 2700, 2600, 2900, 3100, 3500, 3300]

plt.figure(figsize=(10, 5))

# Plot and fill Spotify area
plt.plot(days, spotify_users, color="green", label="Spotify")
plt.fill_between(days, spotify_users, color="green", alpha=0.4)

# Plot and fill YouTube area
plt.plot(days, youtube_users, color="blue", label="YouTube")
plt.fill_between(days, youtube_users, color="blue", alpha=0.3)

plt.legend()
plt.grid(True, linestyle="--", alpha=0.4)
plt.show()
```

Simulate the average delivery time (in minutes) for Flipkart orders over 10 days, and add error bars to represent the standard deviation for each day's delivery time using Matplotlib's errorbar() function.<br><br><em><strong>Hint:</strong> Use the yerr parameter to show error bars for each data point.</em>

```python
days = list(range(1, 11))

avg_delivery_time = [35, 40, 38, 42, 36, 45, 39, 43, 37, 41]

std_deviation = [3, 4, 2, 5, 3, 4, 2, 3, 4, 2]

plt.figure(figsize=(10, 5))

plt.errorbar(
    days,
    avg_delivery_time,
    yerr=std_deviation,
    fmt="o-",
    color="#0F737C",
    ecolor="#D35454",
    capsize=5,
    label="Average Delivery Time"
)
plt.legend()
plt.show()
```

Plot a time series line chart showing your daily Paytm wallet balance over the last 14 days using Matplotlib. Format the x-axis to display dates in 'DD MMM' format (e.g., '12 Jun').

```python
dates = pd.date_range(end=pd.Timestamp.today().normalize(), periods=14)

# Simulated Paytm wallet balances (₹)
balance = [2500, 2200, 2800, 2600, 3100, 2900, 3500,
           3200, 3800, 3600, 4000, 3700, 4300, 4500]

# Create the time series line chart
plt.figure(figsize=(10, 5))
plt.plot(dates, balance, marker='o', color="#B12A08", linestyle='-')

# Format x-axis dates as DD MMM
plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%d %b'))

plt.grid(True, linestyle='--')
plt.show()
```

Refactor your area chart code to add gridlines and custom axis titles, making the chart look more like the analytics dashboards you see in apps like Instagram Insights.<br><br><em><strong>Constraint:</strong> Use ax.set_title(), ax.set_xlabel(), ax.set_ylabel(), and ax.grid(True) in your code.</em>

```python
days = list(range(1, 8))
profile_visits = [120, 150, 135, 180, 220, 200, 260]

fig, ax = plt.subplots(figsize=(8,4))

ax.fill_between(days, profile_visits, color='purple', alpha=0.4)
ax.plot(days, profile_visits, color='purple', marker='o')

# Add title and custom axis labels
ax.set_title('Instagram Insights: Daily Profile Visits')
ax.set_xlabel('Day')
ax.set_ylabel('Number of Profile Visits')

# Add gridlines
ax.grid(True)

plt.tight_layout()
plt.show()
```
